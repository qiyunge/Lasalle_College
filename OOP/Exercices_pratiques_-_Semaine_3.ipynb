```python
#ex3_5
s = "abcdefgh"
s_p = ""
for index,char in enumerate(s):
    if index %2 == 0:
        s_p += char
print(s_p)
##
print("version of more efficient:")
s_p =[]
for index,char in enumerate(s):
    if index %2 ==0:
        s_p.append(char)
s_p = "".join(s_p)
print(s_p)
```
