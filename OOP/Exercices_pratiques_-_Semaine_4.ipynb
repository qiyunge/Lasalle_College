```python
#ex4_3
points =[(2,3),(5,1),(0,4),(3,3)]
def point_the_farest(points:list) -> tuple:
    max_distance = -1
    result = ()
    for point in points:
        distance_square = point[0]**2 + point[1]**2
        if distance_square > max_distance:
            max_distance = distance_square
            result = point
    return result

print(point_the_farest(points))

    
```

```python
#ex4_8
grades ={
    "Alice":"A",
    "Bob":"B",
    "Charlie":"A",
    "Diana":"C",
}

students_regroup ={}
for name,grade in grades.items():
    if grade not in students_regroup.keys():
        students_regroup[grade] = [name]
    else:
        students_regroup[grade].append(name)
print(students_regroup)

#EX4_9
coordinates =[(1,2),(3,4),(1,2),(5,6)]
count_coordinates ={}
for coord in coordinates:
    if coord not in count_coordinates.keys():
        count_coordinates[coord] =1
    else:
        count_coordinates[coord] +=1
print(count_coordinates)
```

```python
#ex4_10
file_path = "./data/ex4_4_output.txt"
file_output_path = "./data/ex4_10_output.txt"
def process_numbers(file_path:str, file_output_path:str) -> None:
    with open(file_path,"r") as f:
        nums_positive = (n for line in f if (n := int(line.strip("\n"))) > 0)
        with open(file_output_path,"w") as fout:
            for num in nums_positive:
                fout.write(f"{num**2}\n")

process_numbers(file_path,file_output_path)
```
