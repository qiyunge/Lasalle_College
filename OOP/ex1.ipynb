```python
# ex1_3
num1 = input("Input first number:")
num2 = input("Input second number:")
num3 = input("Input third number:")

is_num1_number = num1.lstrip("-").isdigit()
is_num2_number = num2.lstrip("-").isdigit()
is_num3_number = num3.lstrip("-").isdigit()

if not(is_num1_number and is_num2_number and is_num3_number):
    print("One of the inputs is not a number.")
else:
    max_num = max(int(num1),int(num2),int(num3))
    print(f"The maximum number is: {max_num}")  
```
