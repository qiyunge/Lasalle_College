```python
from sklearn.datasets import make_regression
```

```python
X, y = make_regression(n_samples=10000,
                       
                        noise=100,
                        random_state=random_state)

type(X), type(y)
X.shape, y.shape
```

```python
# convert the distrubution to a skewed distribution
y_skewed = np.expm1((y +np.abs(np.min(y)))/200)
# y = np.expm1((y+abs(y.min()))/200)
# y = np.expm1((y+abs(y.min()))/200)
```

```python
plt.hist(y_skewed, bins=50)
# plt.hist(y, bins=50)
# px.histogram(y_skewed, nbins=50, title='Skewed Distribution')
```

# transform the skewed distribution to a normal distribution

```python
y_skewed = np.log1p(y_skewed)
```

```python
plt.hist(y_skewed, bins=50)
```

```python
y_origin = np.expm1(y_skewed)
```

```python
plt.hist(y_origin, bins=50)
```
