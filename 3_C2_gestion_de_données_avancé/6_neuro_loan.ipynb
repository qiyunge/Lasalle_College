```python
ms = MinMaxScaler(clip=True)
columns_to_scale = ["ApplicantIncome", "CoapplicantIncome", "LoanAmount","Loan_Amount_Term"]
ms.fit(df[columns_to_scale])
df[columns_to_scale] = ms.transform(df[columns_to_scale])

df.head()
```
