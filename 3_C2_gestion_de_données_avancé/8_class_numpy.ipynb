```python
img = Image.open("./data/Lenna.png")
img
```

```python
img = img.resize((256,256), resample=Image.BICUBIC)
img
```

```python
type(img)
```

```python
arr = np.asarray(img)
# arr = np.array(img)
arr.shape
```

```python
plt.imshow(arr)
```

```python
cmaps = ["Reds", "Blues", "Greens"]
for c in range(3):
    arr_c = arr[:,:,c]
    plt.imshow(arr_c, cmap=cmaps[c], vmin=0, vmax=255)
    plt.show()
```

```python
image_array_2 = arr.copy()
image_array_2[:,:,0] = image_array_2[:,:,0] * 0.6

fig, ax = plt.subplots(1, 2, figsize=(10,5))
ax[0].imshow(image_array_2[:,:,0], cmap="Reds", vmin=0, vmax=255)
ax[1].imshow(arr[:,:,0], cmap="Reds", vmin=0, vmax=255)

ax[0].set_title("Red channel with 0.6 factor")
ax[1].set_title("Red channel original")
plt.show()
```

```python
type(img)
img_face = img.crop((100,100,180,220))
plt.imshow(img_face,vmin=0, vmax=255)
```

```python
img_face_rot = img_face.rotate(45)
plt.imshow(img_face_rot,vmin=0, vmax=255)
```

```python
img_gray = arr.mean(axis=2)
# img_gray = np.average(arr, axis=2, keepdims=True)
# img_gray = img_gray.squeeze(axis = -1)
plt.imshow(img_gray, cmap="gray", vmin=0, vmax=255)
```

```python
img_gray.shape
```

```python
img_min_max  = arr.copy()
img_min_max[:,:,0] = ( np.max(arr[:,:,:3], axis=2) + np.min(arr[:,:,:3], axis=2) ) / 2
img_min_max[:,:,1] = img_min_max[:,:,0]
img_min_max[:,:,2] = img_min_max[:,:,0]
plt.imshow(img_min_max, cmap="gray", vmin=0, vmax=255)
```

```python
img_luma = arr.copy()
img_luma[:,:,0] =  np.dot(arr[:,:,:3], [0.299, 0.587, 0.114])
img_luma[:,:,1] = img_luma[:,:,0]
img_luma[:,:,2] = img_luma[:,:,0]
plt.imshow(img_luma, cmap="gray", vmin=0, vmax=255)
```

```python
img_min = arr.copy()
img_min[:,:,0] = np.min(arr[:,:,:3], axis=2)
img_min[:,:,1] = img_min[:,:,0]
img_min[:,:,2] = img_min[:,:,0]
```

```python
img_max = arr.copy()
img_max[:,:,0] = np.max(arr[:,:,:3], axis=2)
img_max[:,:,1] = img_max[:,:,0]
img_max[:,:,2] = img_max[:,:,0]
```

```python
fig, ax = plt.subplots(2, 3, figsize=(9,5))
ax[0,0].imshow(img_gray, cmap="gray", vmin=0, vmax=255)
ax[0,0].set_title("Average")
ax[0,1].imshow(img_min_max, cmap="gray", vmin=0, vmax=255)
ax[0,1].set_title("Min-Max Average")
ax[0,2].imshow(img_luma, cmap="gray", vmin=0, vmax=255)
ax[0,2].set_title("Luma")
ax[1,0].imshow(img_min, cmap="gray", vmin=0, vmax=255)
ax[1,0].set_title("Min")
ax[1,1].imshow(img_max, cmap="gray", vmin=0, vmax=255)
ax[1,1].set_title("Max")
ax[1,2].imshow(arr[:,:,:],vmin=0, vmax=255)
ax[1,2].set_title("Original")
plt.tight_layout()
plt.show()
```
