## Partie 1 : Tableaux - Approfondissement

### Opérations et Modèles sur les Tableaux

Les tableaux sont la structure de données la plus fondamentale. Maîtrisons-les !

````python
print("start")
````

````python
# Opérations de base sur les tableaux
arr = [1, 2, 3, 4, 5]

# Accès par index (O(1))
print(f"Premier élément : {arr[0]}")
print(f"Dernier élément : {arr[-1]}")
print(f"Élément du milieu : {arr[len(arr)//2]}")

# Découpage (O(k) où k est la taille de la tranche)
print(f"Premiers 3 : {arr[:3]}")
print(f"Derniers 2 : {arr[-2:]}")
print(f"Milieu : {arr[1:4]}")
print(f"Inversé : {arr[::-1]}")

# Opérations courantes
arr.append(6)        # O(1) amorti
arr.insert(0, 0)    # O(n) - décale les éléments
arr.pop()            # O(1) - supprime le dernier
arr.pop(0)           # O(n) - décale les éléments
print(f"Après opérations : {arr}")
````

````python
# Modèles de parcours
arr = [10, 20, 30, 40, 50]

# Parcours avant
print("Avant :")
for i in range(len(arr)):
    print(f"Index {i} : {arr[i]}")

# Parcours arrière
print("\nArrière :")
for i in range(len(arr) - 1, -1, -1):
    print(f"Index {i} : {arr[i]}")

# Deux pointeurs (des deux extrémités)
print("\nDeux pointeurs des extrémités :")
left, right = 0, len(arr) - 1
while left < right:
    print(f"Gauche : {arr[left]}, Droite : {arr[right]}")
    left += 1
    right -= 1
````

````python
# Recherche d'éléments
arr = [3, 7, 1, 9, 2, 5]

# Recherche linéaire - O(n)
def linear_search(arr, target):
    for i, num in enumerate(arr):
        if num == target:
            return i
    return -1

print(f"Index de 9 : {linear_search(arr, 9)}")
print(f"Index de 10 : {linear_search(arr, 10)}")

# Recherche max/min - O(n)
print(f"Max : {max(arr)}, Min : {min(arr)}")
print(f"Index max : {arr.index(max(arr))}")
````

## Partie 2 : Chaînes

### Manipulation de Chaînes

Les chaînes sont essentiellement des tableaux de caractères. De nombreuses techniques sur tableaux s'appliquent !

````python
# Bases des chaînes
s = "Hello World"

# Accès aux caractères (comme les tableaux)
print(f"Premier caractère : {s[0]}")
print(f"Dernier caractère : {s[-1]}")
print(f"Sous-chaîne : {s[0:5]}")

# Méthodes sur chaînes
print(f"Longueur : {len(s)}")
print(f"Majuscules : {s.upper()}")
print(f"Minuscules : {s.lower()}")
print(f"Diviser : {s.split()}")
print(f"Remplacer : {s.replace('World', 'Python')}")
print(f"Compter 'l' : {s.count('l')}")
print(f"Trouver 'World' : {s.find('World')}")
````

````python
# Parcours de chaîne
s = "python"

# Caractère par caractère
for char in s:
    print(char, end=" ")
print()

# Avec index
for i, char in enumerate(s):
    print(f"Index {i} : {char}")

# Inverser la chaîne
reversed_s = s[::-1]
print(f"Inversé : {reversed_s}")
````

````python
# Modèles courants sur chaînes

# Vérifier si palindrome
def is_palindrome(s):
    s = s.lower().replace(" ", "")  # Normaliser
    return s == s[::-1]

print(f"'racecar' est palindrome : {is_palindrome('racecar')}")
print(f"'hello' est palindrome : {is_palindrome('hello')}")

# Compter la fréquence des caractères
def count_chars(s):
    freq = {}
    for char in s:
        freq[char] = freq.get(char, 0) + 1
    return freq

print(f"Fréquence : {count_chars('hello')}")
````

## Partie 3 : Tables de Hachage (Dictionnaires)

### Comprendre les Tables de Hachage

**Table de Hachage** (Dictionnaire en Python) = Stockage clé-valeur rapide avec opérations O(1) en moyenne !

**Pensez-y comme :** Un annuaire téléphonique - vous connaissez le nom (clé), trouvez instantanément le numéro (valeur) !

**Comment ça fonctionne (Étape par Étape) :**
```
Étape 1 : Stocker "Alice" → 95
  "Alice" → Fonction de Hachage → Index 2 → Stocker 95 à l'index 2

Étape 2 : Stocker "Bob" → 87
  "Bob" → Fonction de Hachage → Index 0 → Stocker 87 à l'index 0

Étape 3 : Rechercher "Alice"
  "Alice" → Fonction de Hachage → Index 2 → Retourner 95 (O(1) !)
```

**Exemple Visuel :**
```
Table de Hachage : {"Alice": 95, "Bob": 87, "Charlie": 92}
         ↓
    [Index]  [Clé]      [Valeur]
      0      "Bob"   →   87
      1      None    →   None
      2      "Alice" →   95
      3      "Charlie" → 92
```

**Pourquoi O(1) ?**
- La fonction de hachage calcule instantanément l'index à partir de la clé
- Accès direct au tableau = temps constant
- Pas besoin de chercher parmi toutes les clés !

**Concept Clé :** La fonction de hachage convertit la clé → index du tableau pour un accès instantané !

**Qu'est-ce qu'une fonction de hachage ?** C'est comme une boîte magique qui prend un nom et vous donne un numéro de boîte !

**Exemple Simple :**
- Vous voulez stocker "Alice" → 95
- Fonction de hachage : "Alice" → Boîte #2
- Vous mettez 95 dans la Boîte #2
- Plus tard, quand vous demandez "Où sont les données d'Alice ?"
- Fonction de hachage : "Alice" → Boîte #2 (même boîte !)
- Vous allez instantanément à la Boîte #2 et obtenez 95

**Pourquoi c'est rapide ?** Au lieu de vérifier chaque boîte (Boîte 0, Boîte 1, Boîte 2...), la fonction de hachage vous dit exactement quelle boîte regarder ! Python fait cela automatiquement pour vous !

````python
# Bases des Tables de Hachage
# Créer une table de hachage (dictionnaire) pour stocker les notes des étudiants
# Format : {clé : valeur} où clé est le nom de l'étudiant, valeur est la note
student_grades = {"Alice": 95, "Bob": 87, "Charlie": 92}

# Toutes ces opérations sont O(1) en moyenne - super rapide !

# 1. Recherche : Obtenir la valeur par clé
# Comment ça fonctionne : La fonction de hachage convertit "Alice" → index → obtenir la valeur instantanément
print(f"Note d'Alice : {student_grades['Alice']}")  # O(1) - recherche instantanée !

# 2. Insertion : Ajouter une nouvelle paire clé-valeur
# Comment ça fonctionne : La fonction de hachage convertit "Diana" → index → stocker la valeur à cet index
student_grades["Diana"] = 88  # O(1) - insertion instantanée !

# 3. Mise à jour : Changer la valeur pour une clé existante
# Comment ça fonctionne : Même chose que l'insertion - la fonction de hachage trouve l'index, met à jour la valeur là
student_grades["Bob"] = 90    # O(1) - mise à jour instantanée !

# 4. Suppression : Supprimer une paire clé-valeur
# Comment ça fonctionne : La fonction de hachage trouve l'index, le marque comme vide
del student_grades["Charlie"]  # O(1) - suppression instantanée !

print(f"Mis à jour : {student_grades}")

# 5. Vérifier l'existence : La clé est-elle dans le dictionnaire ?
# Comment ça fonctionne : La fonction de hachage vérifie si la clé existe à l'index calculé
print(f"'Alice' dans dict : {'Alice' in student_grades}")  # O(1) - vérification instantanée !

# 6. Obtenir avec valeur par défaut : Obtenir la valeur ou retourner la valeur par défaut si la clé n'existe pas
# Pourquoi utile ? Empêche KeyError si la clé n'existe pas
# .get(clé, défaut) retourne la valeur par défaut si la clé n'est pas trouvée
print(f"Obtenir avec défaut : {student_grades.get('Eve', 'Non trouvé')}")  # O(1)
````

### Le Modèle du Problème Two Sum

**Problème :** Trouver deux nombres qui s'additionnent pour donner la cible.

**L'Idée :** Au lieu de vérifier toutes les paires (O(n²)), stocker ce qu'on a vu et vérifier si le complément existe !

**Parcours Étape par Étape :**
```
Tableau : [2, 7, 11, 15], cible = 9
vu = {}  (stocke : valeur → index)

i=0, num=2:
  complément = 9 - 2 = 7
  Est-ce que 7 est dans vu ? Non → Ajouter 2 à vu : {2: 0}

i=1, num=7:
  complément = 9 - 7 = 2
  Est-ce que 2 est dans vu ? Oui ! → Paire trouvée : [vu[2], 1] = [0, 1] ✓
  
Réponse : indices [0, 1] → valeurs [2, 7] = 9
```

**Pourquoi Table de Hachage ?**
- Sans table de hachage : Vérifier chaque paire → O(n²) temps
- Avec table de hachage : Un seul passage, recherche O(1) → O(n) temps !

**Idée Clé :** Pour chaque nombre, nous avons besoin de son complément (cible - num). Stocker les nombres au fur et à mesure !

**Qu'est-ce que le "complément" ?** 
- Le complément est "l'autre nombre dont nous avons besoin pour compléter la paire"
- Si cible = 9 et nous voyons num = 2, alors complément = 9 - 2 = 7
- Nous cherchons 7 car 2 + 7 = 9 ✓
- Si nous avons déjà vu 7 dans notre table de hachage, nous avons trouvé notre paire !
- Pensez-y comme : "J'ai 2, j'ai besoin de 7 pour faire 9. Ai-je vu 7 avant ?"

````python
# Modèles courants de tables de hachage

# Modèle 1 : Comptage de fréquence
# Compter combien de fois chaque valeur apparaît
def count_frequency(arr):
    """
    Compter la fréquence de chaque nombre dans le tableau.
    
    Exemple : [1, 2, 2, 3, 3, 3] → {1: 1, 2: 2, 3: 3}
    
    Pourquoi utiliser .get() ? 
    - Si la clé n'existe pas, .get(clé, 0) retourne 0 (défaut)
    - Ensuite nous ajoutons 1 pour obtenir fréquence = 1
    - Si la clé existe, .get(clé, 0) retourne le compte actuel
    - Ensuite nous ajoutons 1 pour incrémenter la fréquence
    """
    freq = {}  # Table de hachage : nombre → compte
    
    for num in arr:
        # Obtenir le compte actuel (0 si première fois qu'on voit ce nombre)
        # Ensuite ajouter 1 pour incrémenter le compte
        freq[num] = freq.get(num, 0) + 1
    
    return freq

arr = [1, 2, 2, 3, 3, 3, 4]
print(f"Fréquence : {count_frequency(arr)}")
# Sortie : {1: 1, 2: 2, 3: 3, 4: 1}
# Signification : 1 apparaît 1 fois, 2 apparaît 2 fois, 3 apparaît 3 fois, 4 apparaît 1 fois

# Modèle 2 : Mapping d'index
# Mapper chaque valeur à son index (ou dernier index si doublons)
def create_index_map(arr):
    """
    Créer un mapping de la valeur vers son index.
    
    Exemple : ['a', 'b', 'c'] → {'a': 0, 'b': 1, 'c': 2}
    
    Pourquoi utile ? Recherche O(1) pour trouver l'index d'une valeur !
    Sans table de hachage : O(n) recherche linéaire pour trouver l'index
    """
    index_map = {}  # Table de hachage : valeur → index
    
    # enumerate nous donne des paires (index, valeur)
    for i, val in enumerate(arr):
        # Stocker : valeur → son index
        # Si la valeur apparaît plusieurs fois, cela stocke le DERNIER index
        index_map[val] = i
    
    return index_map

print(f"Map d'index : {create_index_map(['a', 'b', 'c'])}")
# Sortie : {'a': 0, 'b': 1, 'c': 2}
# Maintenant on peut faire : index_map['b'] → obtenir instantanément 1 (O(1) au lieu de O(n) !)
````

````python
# Le Modèle du Problème Two Sum
# Trouver deux nombres qui s'additionnent pour donner la cible

def two_sum(arr, target):
    """
    Trouver deux nombres qui s'additionnent pour donner la cible.
    
    Temps : O(n) - un seul passage dans le tableau
    Espace : O(n) - la table de hachage stocke jusqu'à n éléments
    
    Sans table de hachage : O(n²) - il faudrait des boucles imbriquées pour vérifier toutes les paires
    
    Stratégie :
    1. Pour chaque nombre, calculer ce dont nous avons besoin (complément = cible - num)
    2. Vérifier si nous avons vu le complément avant (recherche O(1) !)
    3. Si oui, nous avons trouvé notre paire !
    4. Si non, stocker le nombre actuel pour les recherches futures
    
    Exemple : arr = [2, 7, 11, 15], cible = 9
    - i=0, num=2: complément=7, pas dans vu → stocker {2: 0}
    - i=1, num=7: complément=2, EST dans vu ! → retourner [0, 1] ✓
    """
    seen = {}  # Table de hachage : valeur → index (stocke les nombres qu'on a vus jusqu'à présent)
    
    # Parcourir le tableau une fois
    for i, num in enumerate(arr):
        # Calculer quel nombre nous devons associer au nombre actuel
        complement = target - num
        
        # Vérifier si nous avons vu le complément avant (recherche O(1) !)
        if complement in seen:
            # Trouvé ! Retourner les indices du complément et du nombre actuel
            return [seen[complement], i]
        
        # N'avons pas encore vu le complément - stocker le nombre actuel pour plus tard
        # Stocker : nombre → son index (pour pouvoir retourner les indices plus tard)
        seen[num] = i
    
    # Aucune paire trouvée
    return []

# Test
arr = [2, 7, 11, 15]
target = 9
result = two_sum(arr, target)
print(f"Indices Two Sum : {result}")
print(f"Valeurs : {arr[result[0]]}, {arr[result[1]]}")
````

### Collisions de Hachage (Concept)

**Collision :** Quand deux clés différentes hachées donnent le même index.

**Solutions :**

- Adressage ouvert : Trouver le prochain emplacement disponible

## Part 4: Sets

### Opérations sur les Ensembles

**Ensembles** sont comme les tables de hachage mais stockent uniquement les clés (pas de valeurs). Parfait pour :
- Vérifier l'appartenance (O(1))
- Supprimer les doublons
- Opérations sur ensembles (union, intersection, différence)

````python
# Bases des Ensembles
numbers = {1, 2, 3, 4, 5}

# Vérification d'appartenance O(1)
print(f"3 dans l'ensemble : {3 in numbers}")
print(f"10 dans l'ensemble : {10 in numbers}")

# Ajouter/supprimer
numbers.add(6)      # O(1)
numbers.remove(1)   # O(1)
print(f"Après opérations : {numbers}")
````

````python
# Opérations sur Ensembles
set_a = {1, 2, 3, 4, 5}
set_b = {4, 5, 6, 7, 8}

print(f"Ensemble A : {set_a}")
print(f"Ensemble B : {set_b}")
print(f"Union (A | B) : {set_a | set_b}")
print(f"Intersection (A & B) : {set_a & set_b}")
print(f"Différence (A - B) : {set_a - set_b}")
print(f"Différence Symétrique (A ^ B) : {set_a ^ set_b}")
````

````python
# Modèles courants sur ensembles

# Supprimer les doublons d'une liste
arr = [1, 2, 2, 3, 3, 3, 4]
unique = list(set(arr))
print(f"Unique : {unique}")

# Vérifier les doublons
def has_duplicates(arr):
    return len(arr) != len(set(arr))

print(f"A des doublons : {has_duplicates([1, 2, 3])}")
print(f"A des doublons : {has_duplicates([1, 2, 2, 3])}")
````

## Partie 5 : Quand Utiliser Quoi ?

| Structure de Données | Utiliser Quand | Complexité Temporelle |
|----------------------|----------------|----------------------|
| **Tableau** | Besoin de données ordonnées, accès par index | Accès : O(1), Recherche : O(n) |
| **Table de Hachage** | Besoin de paires clé-valeur, recherche rapide | Toutes opérations : O(1) moy |
| **Ensemble** | Besoin d'unicité, vérifications d'appartenance | Toutes opérations : O(1) moy |

**Arbre de Décision :**
- Besoin de recherche rapide par clé ? → Table de Hachage
- Besoin de vérifier si un élément existe ? → Ensemble
- Besoin de données ordonnées/indexées ? → Tableau
- Besoin de compter les fréquences ? → Table de Hachage

```python
def group_anagrams(strs):
    groups = {}
    for s in strs:
        key = ''.join(sorted(s))
        if key not in groups:
            groups[key] = []
        groups[key].append(s)
    return list(groups.values())
```

### Solution 4: Longest Substring (Sliding Window)
```python
def length_of_longest_substring(s):
    char_set = set()
    left = 0
    max_len = 0
    
    for right in range(len(s)):
        while s[right] in char_set:
            char_set.remove(s[left])
            left += 1
        char_set.add(s[right])
        max_len = max(max_len, right - left + 1)
    
    return max_len
```
