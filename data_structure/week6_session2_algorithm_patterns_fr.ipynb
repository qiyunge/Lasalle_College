## Partie 1 : Technique des deux pointeurs

### Motif des deux pointeurs

**Deux pointeurs** = Utiliser deux pointeurs qui se déplacent dans la structure de données.

**Motif 1 : Extrémités opposées**
```
Tableau : [2, 7, 11, 15], cible = 9
          ↑              ↑
         gauche        droite
         (2)           (15)
         2+15=17 > 9, déplacer droite ←
         2+11=13 > 9, déplacer droite ←
         2+7=9 ✓ Trouvé !
```

**Motif 2 : Rapide et lent (Tortue et lièvre)**

**L'idée :** Déplacer deux pointeurs à des vitesses différentes. Quand le rapide atteint la fin, le lent est au milieu !

**Marche pas à pas :**
```
Initial : 1 → 2 → 3 → 4 → 5
          ↑
    lent,rapide (tous deux commencent au nœud 1)

Étape 1 : Déplacer lent +1, rapide +2
         1 → 2 → 3 → 4 → 5
             ↑   ↑
           lent rapide

Étape 2 : Déplacer lent +1, rapide +2
         1 → 2 → 3 → 4 → 5
                 ↑       ↑
               lent     rapide

Étape 3 : Rapide ne peut plus bouger (atteint la fin), stop !
         1 → 2 → 3 → 4 → 5
                 ↑
               lent = milieu ! ✓
```

**Pourquoi ça fonctionne :**
- Le rapide se déplace 2× plus vite que le lent (comme une voiture de course vs un vélo !)
- Quand le rapide parcourt toute la longueur (n nœuds), le lent parcourt la moitié (n/2 nœuds)
- Donc le lent finit au milieu !

**Exemple concret :**
- Liste avec 5 nœuds : 1 → 2 → 3 → 4 → 5
- Rapide se déplace : 1 → 3 → 5 (2 pas à chaque fois, atteint la fin)
- Lent se déplace : 1 → 2 → 3 (1 pas à chaque fois, finit au milieu !)
- Résultat : Lent est au nœud 3, qui est le milieu ✓

**Cas limites :**
- Longueur paire : `1 → 2 → 3 → 4` → lent finit au nœud 2 (premier milieu)
- Nœud unique : `1` → lent = rapide = nœud 1

**Cas d'utilisation :**
- Problèmes de tableaux triés
- Vérification de palindrome
- Recherche de paires
- Détection de cycles

````python
# Motif 1 : Extrémités opposées (Tableau trié)
# Two Sum dans un tableau trié

def two_sum_sorted(arr, target):
    """
    Trouver deux nombres dont la somme égale la cible
    Le tableau est trié
    input: arry = (a1,a2,...,an) st a1 <= a2 <= ... <= an, target
    output: indices (i,j) st arr[i] + arr[j] = target, i < j
    naive: O(n^2) en testant toutes les paires

    常见优化分析框架：

    1. 单循环 + 单调约束  
    - 适用于有序结构 / 双指针 / sliding window
    - 通过单调性排除一片搜索区域

    2. 单循环 + 查询  
    - 适用于“遍历一个元素，快速查询另一个条件”
    - 常用结构：hash / binary search / BST

    3. 单循环 + 单调支配剪枝（优化问题）  
    - 适用于存在目标函数与支配关系的问题
    - 通过删除不可能最优的状态减少搜索空间

    4. 单循环 + 贪心（优化问题）  
    - 适用于存在可证明安全的局部最优选择
    - 需要贪心正确性证明

    5. 单循环 + DP  
    - 适用于问题可压缩为状态递推
    - 通过最优子结构和状态转移避免重复计算

    state: left, right,
    """
    left, right = 0, len(arr) - 1

    while left < right:
        current_sum = arr[left] + arr[right]

        if current_sum == target:
            return [left, right]
        elif current_sum < target:
            left += 1  # Besoin d'une somme plus grande
        else:
            right -= 1  # Besoin d'une somme plus petite

    return []

# Test
arr = [2, 7, 11, 15]
print(f"Indices Two sum : {two_sum_sorted(arr, 9)}")
````

````python
# Motif 2 : Rapide et lent (Liste chaînée)
# Trouver le milieu d'une liste chaînée

# Fonction utilitaire pour créer un nœud (avec dictionnaire)
def create_node(val, next_node=None):
    """Créer un nœud de liste chaînée (dictionnaire)"""
    return {'val': val, 'next': next_node}

def find_middle(head):
    """Trouver le nœud milieu avec deux pointeurs"""
    slow = fast = head

    while fast and fast['next']:  # Accès à 'next' via la clé dictionnaire
        slow = slow['next']      # Avancer d'1 pas
        fast = fast['next']['next']  # Avancer de 2 pas

    return slow

# Motif 3 : Supprimer les doublons d'un tableau trié
def remove_duplicates(arr):
    """
    Supprimer les doublons sur place, retourner la nouvelle longueur
    """
    if not arr:
        return 0

    write_ptr = 1  # Position pour écrire le prochain élément unique

    for read_ptr in range(1, len(arr)):
        if arr[read_ptr] != arr[read_ptr - 1]:
            arr[write_ptr] = arr[read_ptr]
            write_ptr += 1

    return write_ptr

# Test
arr = [1, 1, 2, 2, 3, 4, 4, 5]
new_len = remove_duplicates(arr)
print(f"Nouvelle longueur : {new_len}")
print(f"Tableau : {arr[:new_len]}")
````

## Partie 2 : Motif de la fenêtre glissante

### Concept de la fenêtre glissante

**Fenêtre glissante** = Maintenir une fenêtre qui glisse à travers le tableau/chaîne.

**Analogie du monde réel :** Comme un gardien qui regarde un flux de caméra de sécurité !
- La caméra affiche une fenêtre de taille fixe de la scène
- Quand la scène bouge, au lieu de réinitialiser la caméra, on fait juste glisser la vue
- On garde la trace de ce qui est dans la vue actuelle sans tout recalculer

**L'idée :** Au lieu de tout recalculer, faire glisser la fenêtre et mettre à jour de manière incrémentale !

**Pourquoi c'est plus rapide ?**
- **Approche naïve :** Pour chaque fenêtre, additionner tous les k nombres depuis zéro
  - Fenêtre 1 : 2+1+5 = 8 (3 additions)
  - Fenêtre 2 : 1+5+1 = 7 (3 additions à nouveau !)
  - Fenêtre 3 : 5+1+3 = 9 (3 additions à nouveau !)
  - Total : 3 additions × 4 fenêtres = 12 opérations ✗
  
- **Fenêtre glissante :** Calculer une fois, puis réutiliser !
  - Fenêtre 1 : 2+1+5 = 8 (calculer une fois)
  - Fenêtre 2 : 8 - 2 + 1 = 7 (soustraire à gauche, ajouter à droite !)
  - Fenêtre 3 : 7 - 1 + 3 = 9 (soustraire à gauche, ajouter à droite !)
  - Total : 3 + 3 = 6 opérations ✓ (50 % plus rapide !)

**Idée clé :** Ne pas recalculer ! Réutiliser la somme précédente en soustrayant ce qui part et en ajoutant ce qui entre !

**Visuel (Taille fixe k=3) :**
```
Tableau : [2, 1, 5, 1, 3, 2]
          [---]              Fenêtre 1 : somme = 2+1+5 = 8
             [---]           Fenêtre 2 : somme = 8-2+1 = 7
                             (soustraire 2 qui part, ajouter 1 qui entre)
                [---]        Fenêtre 3 : somme = 7-1+3 = 9
                             (soustraire 1 qui part, ajouter 3 qui entre)
                   [---]     Fenêtre 4 : somme = 9-5+2 = 6
                             (soustraire 5 qui part, ajouter 2 qui entre)
Somme max = 9

Pensez-y comme un train en mouvement :
- Fenêtre 1 : [2,1,5] → somme = 8
- Fenêtre 2 : [1,5,1] → retirer 2 (gauche), ajouter 1 (entre) → 8-2+1 = 7
- Fenêtre 3 : [5,1,3] → retirer 1 (gauche), ajouter 3 (entre) → 7-1+3 = 9
```
**Idée clé :** Ne pas recalculer ! Soustraire ce qui part, ajouter ce qui entre. O(n) au lieu de O(nk) !

**Visuel (Taille variable - Expansion et réduction) :**
```
Chaîne : "abcabcbb"
        [a]            Fenêtre : {a}, len=1
        [ab]           Expansion : {a,b}, len=2
        [abc]          Expansion : {a,b,c}, len=3 ✓
        [abca]         Expansion : doublon 'a' trouvé !
         [bca]         Réduire gauche jusqu'à 'a' retiré, len=3 ✓
         [bcab]        Expansion : doublon 'b' trouvé !
          [cab]        Réduire gauche jusqu'à 'b' retiré, len=3 ✓
Longueur max = 3
```

**Marche pas à pas (Fenêtre variable) :**
1. **Expansion :** Déplacer le pointeur `droite`, ajouter à la fenêtre
2. **Vérifier :** La fenêtre est-elle valide ? (pas de doublons, somme ≤ cible, etc.)
3. **Réduction :** Si invalide, déplacer le pointeur `gauche` jusqu'à ce que ce soit valide
4. **Mise à jour :** Suivre la meilleure réponse (longueur max, somme min, etc.)

**Types :**
1. **Taille fixe :** La taille de la fenêtre est constante (plus facile !)
2. **Taille variable :** Expansion/réduction selon une condition (plus flexible)

**Cas d'utilisation :**
- Problèmes de sous-chaîne/sous-tableau
- Maximum/minimum dans une fenêtre
- Plus long/plus court sous-tableau avec condition

````python
# Motif 1 : Fenêtre de taille fixe
# Somme maximale d'un sous-tableau de taille k

def max_sum_subarray(arr, k):
    """
    Trouver la somme maximale d'un sous-tableau de taille k
    input: arr=[a1,a2,...,an], k
    output:maximum (a_i + a_{i+1} + ... + a_{i+k-1}) pour 
    """
    if len(arr) < k:
        return 0

    # Calculer la somme de la première fenêtre
    window_sum = sum(arr[:k])
    max_sum = window_sum

    # Faire glisser la fenêtre
    for i in range(k, len(arr)):
        window_sum = window_sum - arr[i - k] + arr[i]
        max_sum = max(max_sum, window_sum)

    return max_sum

# Test
arr = [2, 1, 5, 1, 3, 2]
k = 3
print(f"Somme max du sous-tableau de taille {k} : {max_sum_subarray(arr, k)}")
````

````python
# Motif 2 : Fenêtre de taille variable
# Plus longue sous-chaîne sans caractères répétés

def length_of_longest_substring(s):
    """
    ## Longest Substring Without Repeating Characters

    **Input**  
    s = (s1, s2, ..., sn)

    **Output**  
    maximum |s_i ... s_j|  
    such that all characters in s_i ... s_j are distinct

    **State**  
    left, right, char_set

    **Invariant**  
    The current window s[left..right] contains no repeated characters.

    **Naive approach**  
    O(n^3), by enumerating all substrings and checking duplicates.

    **Optimization analysis**  
    This problem can be optimized using:

    - single scan of `right`
    - monotone window constraint
    - hash-based membership query
    - variable sliding window

    The constraint “all characters are distinct” is monotone with respect to shrinking the window from the left.  
    Therefore, when a duplicate appears, we can restore feasibility by moving `left` forward.

    **Complexity**  
    Time: O(n)  
    Space: O(min(n, charset size))
    """
    char_set = set()
    left = 0
    max_len = 0

    for right in range(len(s)):
        # Réduire la fenêtre jusqu'à ce qu'il n'y ait plus de doublons
        while s[right] in char_set:
            char_set.remove(s[left])
            left += 1

        # Étendre la fenêtre
        char_set.add(s[right])
        max_len = max(max_len, right - left + 1)

    return max_len

# Test
s = "abcabcbb"
print(f"Longueur de la plus longue sous-chaîne : {length_of_longest_substring(s)}")
````

````python
# Motif 3 : Fenêtre de sous-chaîne minimale
# Trouver la fenêtre minimale dans s contenant tous les caractères de t

def min_window(s, t):
    """
    Trouver la sous-chaîne fenêtre minimale contenant tous les caractères de t
    input: s (s1, s2, ..., sn), t (t1, t2, ..., tm)
    output: s[i..j] tel que s[i..j] contient tous les caractères de t et j - i + 1 est minimal
    naive: O(n^2 * m) en testant toutes les sous-chaînes et vérifiant la présence de t
        optimization analysis:  single scan of `right`, monotone window constraint, hash-based frequency query, variable sliding window
        state: left,right,have,need,formed,required
    """
    if not s or not t:
        return ""

    # Compter les caractères dans t
    need = {}
    for char in t:
        need[char] = need.get(char, 0) + 1

    have = {}
    left = 0
    min_len = float('inf')
    min_window_start = 0
    required = len(need)
    formed = 0

    for right in range(len(s)):
        char = s[right]
        have[char] = have.get(char, 0) + 1

        # Vérifier si le caractère actuel satisfait l'exigence
        if char in need and have[char] == need[char]:
            formed += 1

        # Tenter de réduire la fenêtre
        while left <= right and formed == required:
            # Mettre à jour la fenêtre minimale
            if right - left + 1 < min_len:
                min_len = right - left + 1
                min_window_start = left

            # Retirer le caractère le plus à gauche
            left_char = s[left]
            have[left_char] -= 1
            if left_char in need and have[left_char] < need[left_char]:
                formed -= 1

            left += 1

    return "" if min_len == float('inf') else s[min_window_start:min_window_start + min_len]

# Test
s = "ADOBECODEBANC"
t = "ABC"
print(f"Fenêtre min : '{min_window(s, t)}'")
````

## Partie 3 : BFS/DFS pour les graphes

### Quand utiliser BFS vs DFS

**Comparaison visuelle :**
```
Graphe :   0
          / \
         1   2
        / \   \
       3   4   5

Ordre BFS : 0 → 1 → 2 → 3 → 4 → 5  (niveau par niveau)
Ordre DFS : 0 → 1 → 3 → 4 → 2 → 5  (aller en profondeur d'abord)
```

**BFS (Parcours en largeur) :**
- **Comment :** Utiliser une file, explorer niveau par niveau
- **Quand :** Plus court chemin (graphes non pondérés), parcours par niveau
- **Pensez :** « Visiter tous les voisins avant d'aller plus profond »

**DFS (Parcours en profondeur) :**
- **Comment :** Utiliser récursion/pile, aller en profondeur avant de revenir
- **Quand :** Explorer tous les chemins, tri topologique, détection de cycles
- **Pensez :** « Aller le plus profond possible, puis revenir »

**Décision rapide :**
- Besoin du plus court chemin ? → BFS
- Besoin de tous les chemins ou exploration en profondeur ? → DFS

````python
from collections import deque

# BFS : Plus court chemin (Graphe non pondéré)
def shortest_path(graph, start, end):
    """
    Trouver le plus court chemin avec BFS
    """
    if start == end:
        return [start]

    queue = deque([(start, [start])])
    visited = {start}

    while queue:
        node, path = queue.popleft()

        for neighbor in graph[node]:
            if neighbor == end:
                return path + [neighbor]

            if neighbor not in visited:
                visited.add(neighbor)
                queue.append((neighbor, path + [neighbor]))

    return []  # Pas de chemin trouvé

# Test
graph = {
    0: [1, 2],
    1: [0, 3],
    2: [0, 3],
    3: [1, 2, 4],
    4: [3]
}

print(f"Plus court chemin 0->4 : {shortest_path(graph, 0, 4)}")
````

````python
# DFS : Tous les chemins entre deux nœuds
def all_paths(graph, start, end):
    """
    Trouver tous les chemins avec DFS
    """
    paths = []

    def dfs(node, path, visited):
        if node == end:
            paths.append(path[:])
            return

        for neighbor in graph[node]:
            if neighbor not in visited:
                visited.add(neighbor)
                path.append(neighbor)
                dfs(neighbor, path, visited)
                path.pop()
                visited.remove(neighbor)

    dfs(start, [start], {start})
    return paths

print(f"Tous les chemins 0->3 : {all_paths(graph, 0, 3)}")
````

````python
# Exemple : Appliquer le cadre
# Problème : Trouver tous les anagrammes dans une chaîne

# Rappel : Les anagrammes sont des mots avec les mêmes lettres dans un ordre différent.
# Exemple : "abc" et "bac" sont des anagrammes (tous deux ont a, b, c)
# Dans la chaîne "cbaebabacd", trouver toutes les sous-chaînes qui sont des anagrammes de "abc"
# Réponse : "cba" (index 0) et "bac" (index 6) sont des anagrammes de "abc"

def find_anagrams(s, p):
    """
    Trouver tous les indices de début des anagrammes de p dans s

    Étape 1 : Comprendre
    - Entrée : chaîne s, chaîne p
    - Sortie : liste des indices de début
    - Contrainte : p.length <= s.length

    Étape 2 : Choisir la SD
    - Besoin de comptage de fréquence → Table de hachage

    Étape 3 : Choisir l'algorithme
    - Fenêtre fixe de taille len(p) → Fenêtre glissante

    Étape 4 : Cas limites
    - len(p) > len(s) → return []
    - Chaînes vides
    """
    if len(p) > len(s):
        return []

    # Fréquence de p
    p_freq = {}
    for char in p:
        p_freq[char] = p_freq.get(char, 0) + 1

    # Fenêtre glissante
    window_freq = {}
    result = []

    # Initialiser la fenêtre
    for i in range(len(p)):
        window_freq[s[i]] = window_freq.get(s[i], 0) + 1

    # Vérifier la première fenêtre
    if window_freq == p_freq:
        result.append(0)

    # Faire glisser la fenêtre
    for i in range(len(p), len(s)):
        # Retirer le plus à gauche
        left_char = s[i - len(p)]
        window_freq[left_char] -= 1
        if window_freq[left_char] == 0:
            del window_freq[left_char]

        # Ajouter le plus à droite
        right_char = s[i]
        window_freq[right_char] = window_freq.get(right_char, 0) + 1

        # Vérifier si anagramme
        if window_freq == p_freq:
            result.append(i - len(p) + 1)

    return result

# Test
s = "cbaebabacd"
p = "abc"
print(f"Indices des anagrammes : {find_anagrams(s, p)}")
````

### Problème 1 : Récipient avec le plus d'eau
Étant donné un tableau de hauteurs, trouver deux lignes qui forment le récipient avec le plus d'eau.

**Indice :** Deux pointeurs aux deux extrémités

````python
# Votre solution ici :
def max_area(height):
    # TODO : À implémenter
    '''
    input:sequence, total ordered
    j() = (j - i) * min(height[i], height[j])
     - i = 0, j = len(height) - 1
    
    当h[i] < h[j]时, 固定i，移动j--， 必然更小，可以剪枝，变为单路径遍历
    '''
    left, right = 0, len(height)-1
    best = 0
    while left < right:
        area = (right - left+1)*min(height[left], height[right])
        if area > best:
            best = area
        if height[left] >= height[right]:
            right -= 1
        else:
            left += 1

    return best

height = [1,8,6,2,5,4,8,3,7]
print(f"Max area : {max_area(height)}")
````

### Problème 2 : Remplacement de caractère répété le plus long
Étant donné une chaîne et k, trouver la plus longue sous-chaîne avec le même caractère après avoir remplacé au plus k caractères.

**Indice :** Fenêtre glissante avec carte de fréquence

````python
# Votre solution ici :
from collections import defaultdict

def character_replacement(s, k):
   """
   input : sequence, totoal ordered,  output : sous-chaine st. predict (right - left +1) - max_freq <= k


   """
   freq = defaultdict(int)
   
   left = 0
   max_freq = 0
   max_len = 0
   for right in range(len(s)):
      freq[s[right]] += 1
      max_freq = max(max_freq, freq[s[right]])
      
     
      if right -left +1 -max_freq > k:
         freq[s[left]] -= 1
         left += 1
         
      max_len = max(max_len, right - left + 1)
   return max_len

test = "AABABBA"
k = 1
print(f"Max length : {character_replacement(test, k)}")

````

✅ **Deux pointeurs :**
- Extrémités opposées : tableaux triés, paires
- Rapide et lent : listes chaînées, cycles
- Réduit O(n²) à O(n)

✅ **Fenêtre glissante :**
- Taille fixe : calculer une fois, faire glisser
- Taille variable : expansion/réduction selon condition
- Idéal pour les problèmes de sous-chaîne/sous-tableau

✅ **BFS vs DFS :**
- BFS : Plus court chemin, parcours par niveau
- DFS : Tous les chemins, exploration en profondeur

✅ **Résolution de problèmes :**
- Comprendre → Choisir SD → Choisir algorithme → Cas limites → Coder
- Lire les contraintes attentivement
- Tester avec des exemples
