# Partie 1 : Récursivité et complexité Big-O

## 1.1 Récursivité

| Concept | Définition | Pourquoi c'est important |
|---------|------------|--------------------------|
| **Récursivité** | Une fonction qui s'appelle elle-même pour résoudre des instances plus petites du même problème | Façon élégante d'exprimer diviser-pour-régner et parcours d'arbres/graphes |
| **Cas de base** | Condition d'arrêt qui évite la récursivité infinie | Évite le débordement de pile et les résultats incorrects |
| **Cas récursif** | Appel avec une entrée plus petite ou plus simple | Doit toujours converger vers le cas de base |
| **Pile d'appels** | Mémoire utilisée pour les appels de fonctions actifs | Explique le coût en espace ; une récursivité profonde peut saturer la mémoire |

**Idée clé :** Décomposer un problème en sous-problèmes identiques jusqu'au cas le plus simple.

### Applications
- **Parcours arbre/graphe :** Le DFS s'exprime naturellement par récursivité
- **Backtracking :** Générer permutations, combinaisons, sous-ensembles
- **Diviser pour régner :** Tri fusion, recherche dichotomique (conceptuellement)
- **Parsing :** Arbres d'expression, analyse syntaxique des compilateurs

## 1.2 Big-O : complexité en temps et en espace

**Big-O** décrit comment un algorithme évolue avec la taille de l'entrée (cas le pire). Il répond à : *« Que se passe-t-il quand n devient grand ? »*

### Complexité en temps (du plus rapide au plus lent)

| Notation | Croissance | Exemple | Quand entrée 10→1000 | Analogie concrète |
|----------|------------|---------|----------------------|--------------------|
| **O(1)** | Constante | Accès par indice dans un tableau | Même temps | Interrupteur : on/off en une étape |
| **O(log n)** | Logarithmique | Recherche dichotomique | ~3× plus lent | Dictionnaire : diviser en deux à chaque étape |
| **O(n)** | Linéaire | Une boucle sur le tableau | 100× plus lent | Lire chaque page d'un livre |
| **O(n log n)** | Linéarithmique | Tri fusion | ~100× plus lent | Tri efficace |
| **O(n²)** | Quadratique | Boucles imbriquées | 10 000× plus lent | Comparer chaque paire d'éléments |
| **O(2ⁿ)** | Exponentielle | Fibonacci naïf | Inutilisable | Explosion combinatoire |

### Pourquoi c'est critique
- **Entretiens :** On attend de vous une analyse de la complexité
- **Conception :** O(log n) vs O(n) peut déterminer si une fonctionnalité est réalisable
- **Débogage :** Une lenteur soudaine provient souvent d'un O(n²) inattendu ou pire

```python
# Rappel : exemples de complexité courante
def binary_search(arr, target):  # O(log n) — à chaque étape on divise l'espace par deux
    left, right = 0, len(arr) - 1
    while left <= right:
        mid = (left + right) // 2
        if arr[mid] == target:
            return mid
        if arr[mid] < target:
            left = mid + 1
        else:
            right = mid - 1
    return -1

print(f"Recherche dichotomique de 7 dans [1,3,5,7,9] : indice {binary_search([1,3,5,7,9], 7)}")
```

# Partie 2 : Tableaux, chaînes et tables de hachage / ensembles

## 2.1 Tableaux et chaînes

| Concept | Définition | Importance |
|---------|------------|------------|
| **Tableaux** | Stockage contigu, accès O(1) par indice | Structure la plus basique ; utilisée dans presque tous les algorithmes |
| **Indexation** | `arr[i]` en O(1) | Accès rapide quand on connaît la position |
| **Deux pointeurs** | Deux indices qui parcourent le tableau | Réduire O(n²) à O(n) pour paires, palindromes, sous-tableaux |
| **Chaînes** | Séquences de caractères immuables | Tableaux avec opérations spécifiques au texte |

### Applications
- **Bases de données :** Lignes stockées comme tableaux
- **Image/Vidéo :** Pixels/trames comme tableaux
- **Cache :** Éviction LRU avec structures type tableau
- **Parsing :** Correspondance de chaînes, regex, compilateurs

## 2.2 Tables de hachage et ensembles

| Structure | Idée clé | Opérations | Quand utiliser |
|-----------|----------|------------|----------------|
| **Table de hachage (dict)** | Association clé → valeur | Recherche, insertion, suppression O(1) (en moy.) | Pour un accès rapide par clé |
| **Ensemble (set)** | Collection non ordonnée d'éléments uniques | Test d'appartenance O(1) | Unicité ou « ai-je déjà vu ça ? » |

**Gestion des collisions (conceptuellement) :** Chaînage (liste à chaque bucket) ou adressage ouvert (trouver le prochain emplacement).

### Applications
- **Comptage de fréquence :** Comptage de caractères, fréquence de mots, analyses
- **Two Sum / Complément :** « Existe-t-il un autre élément qui donne la somme ? » → vérification O(1)
- **Déduplication :** Supprimer les doublons, suivre les nœuds visités dans les graphes
- **Cache/Mémoïsation :** Stocker les résultats calculés par clé d'entrée

```python
# Two Sum classique avec table de hachage (O(n) au lieu de O(n²))
def two_sum(nums, target):
    seen = {}  # valeur -> indice
    for i, num in enumerate(nums):
        complement = target - num
        if complement in seen:
            return [seen[complement], i]
        seen[num] = i
    return []

print(f"Two sum [2,7,11,15], cible 9 : {two_sum([2,7,11,15], 9)}")
```

# Partie 3 : Piles, files d'attente et listes chaînées

## 3.1 Piles (LIFO — dernier entré, premier sorti)

| Concept | Définition | Applications |
|---------|------------|----------------|
| **Pile** | Ajouter/supprimer uniquement en haut | Annulation, parenthèses correspondantes, évaluation d'expressions |
| **Opérations** | push, pop, peek — toutes O(1) | |

**Analogie :** Pile d'assiettes — on ajoute et retire uniquement par le haut.

### Applications
- **Parsing :** Correspondance des accolades `{}`, `[]`, `()`
- **Annuler/Rétablir :** Pile d'historique dans les éditeurs
- **Pile d'appels :** Fonctionnement de la récursivité et des appels de fonction
- **Évaluation d'expressions :** Calculatrices en notation polonaise inverse (RPN)

## 3.2 Files d'attente (FIFO — premier entré, premier sorti)

| Concept | Définition | Applications |
|---------|------------|----------------|
| **File** | Ajouter à l'arrière, retirer à l'avant | Ordre équitable, BFS |
| **Opérations** | enqueue, dequeue — O(1) avec deque | |

**Analogie :** File d'attente en magasin — le premier arrivé est servi en premier.

### Applications
- **BFS :** Parcours par niveau, plus court chemin (graphes non pondérés)
- **Planification de tâches :** Files de jobs, files d'impression
- **Gestion d'événements :** Traiter les événements dans l'ordre d'arrivée

## 3.3 Listes chaînées

| Concept | Définition | Importance |
|---------|------------|------------|
| **Nœud** | Valeur + pointeur vers le suivant (et optionnellement le précédent) | Bloc de construction fondamental |
| **Simplement chaînée** | Une direction : next | Plus simple, moins de mémoire |
| **Doublement chaînée** | Deux directions : next, prev | Suppression plus facile, parcours bidirectionnel |
| **Modèles** | Deux pointeurs (rapide/lent), inversion, détection de cycle | Détection de cycle, milieu, palindrome |

### Applications
- **Cache LRU :** Liste doublement chaînée + table de hachage pour éviction O(1)
- **Historique navigateur :** Navigation précédent/suivant
- **Listes d'annulation :** Quand l'ordre compte mais la structure est flexible
- **Niveau bas :** Allocateurs mémoire, systèmes de fichiers

# Partie 4 : Arbres et parcours

## 4.1 Structure d'arbre

| Terme | Signification |
|-------|---------------|
| **Racine** | Nœud du sommet (sans parent) |
| **Nœud** | Données, enfants, parent optionnel |
| **Feuille** | Nœud sans enfants |
| **Profondeur** | Distance depuis la racine |
| **Hauteur** | Plus long chemin du nœud à une feuille |

## 4.2 Arbre binaire de recherche (ABR)

**Propriété :** Pour chaque nœud : `gauche < nœud < droite` (ou ≤ selon la définition)

**Pourquoi c'est important :** Permet recherche, insertion, suppression en O(log n) si équilibré.

## 4.3 Parcours

| Parcours | Ordre | Cas d'usage |
|----------|-------|--------------|
| **Préordre** | Racine → Gauche → Droite | Copier l'arbre, expressions préfixées |
| **Inordre** | Gauche → Racine → Droite | ABR → ordre trié |
| **Postordre** | Gauche → Droite → Racine | Supprimer l'arbre, expressions postfixées |
| **Par niveau (BFS)** | Niveau par niveau | Plus court chemin, traitement en largeur |

### Applications
- **Bases de données :** B-arbres, B+ arbres pour l'indexation
- **Systèmes de fichiers :** Arbres de répertoires
- **Arbres d'expression :** Parser et évaluer des expressions mathématiques
- **IA/ML :** Arbres de décision

```python
# Modèles de parcours DFS simples (conceptuels)
def inorder(node):  # Gauche -> Racine -> Droite → trié pour ABR
    if not node:
        return []
    return inorder(node['left']) + [node['val']] + inorder(node['right'])

def preorder(node):  # Racine -> Gauche -> Droite
    if not node:
        return []
    return [node['val']] + preorder(node['left']) + preorder(node['right'])

print("Inordre = trié pour ABR ; Préordre = racine en premier")
```

# Partie 5 : Tas et graphes

## 5.1 Tas (file à priorité)

| Concept | Définition | Importance |
|---------|------------|------------|
| **Tas-min** | Parent ≤ enfants ; min à la racine | Obtenir le plus petit en O(log n) |
| **Tas-max** | Parent ≥ enfants ; max à la racine | Obtenir le plus grand en O(log n) |
| **Formules d'indices** | Parent : `(i-1)//2` ; Enfants : `2i+1`, `2i+2` | Représentation en tableau d'un arbre complet |

**En Python :** `heapq` — tas-min ; négater pour tas-max.

### Applications
- **Top K :** K plus grands/plus petits éléments
- **Dijkstra :** Plus court chemin dans les graphes pondérés
- **Fusion de K listes triées :** Un tas des têtes de liste
- **Simulation d'événements :** Traiter par priorité (ex. temps)

```python
import heapq

# K plus grands éléments (O(n log k) avec tas-min de taille k)
def k_largest(nums, k):
    return heapq.nlargest(k, nums)

print(f"3 plus grands de [3,2,1,5,6,4] : {k_largest([3,2,1,5,6,4], 3)}")
```

## 5.2 Graphes

| Concept | Définition |
|---------|------------|
| **Sommet (nœud)** | Entité |
| **Arête** | Lien entre deux sommets |
| **Orienté** | Arêtes à sens unique (ex. liens) |
| **Non orienté** | Arêtes bidirectionnelles (ex. amitiés) |
| **Pondéré** | Arêtes avec coûts |

### Représentation

- **Liste d'adjacence :** `{nœud: [voisins]}` — privilégiée pour graphes creux
- **Matrice d'adjacence :** `matrice[i][j] = 1` si arête — privilégiée pour graphes denses

### Applications
- **Réseaux :** Sociaux, transports, informatiques
- **Cartes/Navigation :** Plus court chemin (Dijkstra, BFS pour non pondérés)
- **Dépendances :** Ordre de compilation, prérequis de cours (tri topologique)
- **Recommandations :** « Personnes que vous connaissez peut-être » (BFS/DFS)

# Partie 6 : Modèles algorithmiques

## 6.1 Deux pointeurs

| Modèle | Quand l'utiliser | Exemple |
|--------|------------------|--------|
| **Extrémités opposées** | Tableau trié, paires, palindromes | Two Sum (trié), Container With Most Water |
| **Rapide et lent** | Listes chaînées, cycles | Nœud central, détection de cycle (Floyd) |

**Intérêt :** Réduit souvent O(n²) à O(n).

## 6.2 Fenêtre glissante

| Type | Quand | Exemple |
|------|-------|--------|
| **Taille fixe** | La taille k de la fenêtre est donnée | Somme max d'un sous-tableau de taille k |
| **Taille variable** | Optimiser la fenêtre selon une condition | Plus longue sous-chaîne sans répétition |

**Intérêt :** Éviter de tout recalculer ; O(n) au lieu de O(n·k).

## 6.3 BFS vs DFS

| Algorithme | Utiliser quand | Structure de données |
|------------|----------------|----------------------|
| **BFS** | Plus court chemin (non pondéré), parcours par niveau | File |
| **DFS** | Tous les chemins, exploration complète, cycles | Pile / Récursivité |

**Règle pratique :** Besoin du plus court chemin ? → BFS. Besoin de tous les chemins ou exploration en profondeur ? → DFS.

# Partie 7 : Guide de décision — Quelle structure choisir ?

| Besoin | Choisir |
|--------|--------|
| Accès rapide par clé | **Table de hachage** |
| Éléments uniques / appartenance | **Ensemble** |
| Ordre important, accès par indice | **Tableau** |
| LIFO (annulation, correspondance) | **Pile** |
| FIFO (BFS, planification) | **File d'attente** |
| Insertion/suppression flexible (au milieu) | **Liste chaînée** |
| Accès trié, hiérarchie | **Arbre / ABR** |
| Priorité / top-K | **Tas** |
| Connexions, relations | **Graphe** |
