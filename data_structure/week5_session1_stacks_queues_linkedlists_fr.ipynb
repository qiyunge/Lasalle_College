## Partie 1 : Piles (LIFO - Dernier Entré Premier Sorti)

### Concept de Pile

**Pile** = LIFO (Dernier Entré Premier Sorti) - comme une pile d'assiettes !

**Visuel :**
```
push(3) → push(2) → push(1)
         ┌─────┐
         │  1  │ ← Sommet (dernier ajouté)
         ├─────┤
         │  2  │
         ├─────┤
         │  3  │ ← Fond (premier ajouté)
         └─────┘
pop() retire 1 (sommet), peek() retourne 1
```

**Opérations :**
- `push(item)` - Ajouter au sommet (O(1))
- `pop()` - Retirer du sommet (O(1))
- `peek()` / `top()` - Voir le sommet sans retirer (O(1))
- `is_empty()` - Vérifier si vide (O(1))

**Cas d'utilisation :**
- Correspondance de parenthèses/crochets
- Opérations d'annulation
- Pile d'appels de fonction
- Évaluation d'expressions

````python
# Implémentation de Pile en utilisant une Liste
# Une pile est juste une liste ! On utilise les opérations de liste pour simuler le comportement d'une pile.

# Opérations de pile (en utilisant directement une liste) :
# - push : utiliser append() pour ajouter à la fin (sommet)
# - pop : utiliser pop() pour retirer de la fin (sommet)
# - peek : utiliser [-1] pour voir le dernier élément
# - is_empty : vérifier si la liste est vide
# - size : utiliser len()

# Créer une pile (juste une liste vide !)
stack = []  # C'est notre pile !

# Push : Ajouter un élément au sommet de la pile
def stack_push(stack, item):
    """Ajouter un élément au sommet de la pile"""
    stack.append(item)  # O(1) - ajouter à la fin

# Pop : Retirer et retourner l'élément du sommet
def stack_pop(stack):
    """Retirer et retourner l'élément du sommet"""
    if len(stack) == 0:  # Vérifier si vide
        return None
    return stack.pop()  # O(1) - retirer de la fin

# Peek : Voir l'élément du sommet sans le retirer
def stack_peek(stack):
    """Voir l'élément du sommet sans le retirer"""
    if len(stack) == 0:
        return None
    return stack[-1]  # Retourner le dernier élément

# Is empty : Vérifier si la pile est vide
def stack_is_empty(stack):
    """Vérifier si la pile est vide"""
    return len(stack) == 0

# Size : Obtenir le nombre d'éléments
def stack_size(stack):
    """Obtenir le nombre d'éléments dans la pile"""
    return len(stack)

# Test
stack_push(stack, 1)
stack_push(stack, 2)
stack_push(stack, 3)
print(f"Pile : {stack}")
print(f"Sommet : {stack_peek(stack)}")
print(f"Retiré : {stack_pop(stack)}")
print(f"Pile après pop : {stack}")

# Note : En pratique, vous pouvez utiliser la liste directement !
# stack = []
# stack.append(1)  # push
# stack.pop()     # pop
# stack[-1]       # peek
````

````python
# Problème : Parenthèses Valides
# Vérifier si les parenthèses sont équilibrées : "()[]{}"

def is_valid_parentheses(s):
    stack = []
    pairs = {')': '(', '}': '{', ']': '['}
    
    for char in s:
        if char in pairs:  # Crochet fermant
            top = stack.pop()# if stack else '#'  # Si la pile est vide, utiliser un caractère de remplacement
            if not stack or top != pairs[char]:
                return False
           
        else:  # Crochet ouvrant
            stack.append(char)
    
    return len(stack) == 0

# Test
print(f"'()' valide : {is_valid_parentheses('()')}")
print(f"'()[]{{}}' valide : {is_valid_parentheses('()[]{}')}")
print(f"'(]' valide : {is_valid_parentheses('(]')}")
print(f"'([)]' valide : {is_valid_parentheses('([()()()()()())]')}")
````

## Partie 2 : Files d'attente (FIFO - Premier Entré Premier Sorti)

### Concept de File d'attente

**File d'attente** = FIFO (Premier Entré Premier Sorti) - comme une file dans un magasin !

**Visuel :**
```
enqueue(1) → enqueue(2) → enqueue(3)
Avant → [1] [2] [3] ← Arrière
         ↑           ↑
      dequeue()   enqueue()
      retire 1    ajoute ici
```

**Opérations :**
- `enqueue(item)` - Ajouter à l'arrière (O(1))
- `dequeue()` - Retirer de l'avant (O(1) avec collections.deque, O(n) avec list)
- `front()` / `peek()` - Voir l'avant sans retirer (O(1))
- `is_empty()` - Vérifier si vide (O(1))

**Cas d'utilisation :**
- BFS (Parcours en Largeur)
- Planification de tâches
- File d'impression
- Parcours d'arbre niveau par niveau

````python
# Implémentation de File d'attente en utilisant collections.deque (efficace !)
from collections import deque

# Une file d'attente est un deque ! On utilise directement les opérations de deque.
# deque = "double-ended queue" - efficace pour les deux extrémités

# Opérations de file d'attente (en utilisant directement deque) :
# - enqueue : utiliser append() pour ajouter à l'arrière
# - dequeue : utiliser popleft() pour retirer de l'avant
# - front : utiliser [0] pour voir le premier élément
# - is_empty : vérifier si le deque est vide
# - size : utiliser len()

# Créer une file d'attente (juste un deque !)
queue = deque()  # C'est notre file d'attente !

# Enqueue : Ajouter un élément à l'arrière de la file
def queue_enqueue(queue, item):
    """Ajouter un élément à l'arrière de la file"""
    queue.append(item)  # O(1) - ajouter à la fin

# Dequeue : Retirer et retourner l'élément de l'avant
def queue_dequeue(queue):
    """Retirer et retourner l'élément de l'avant"""
    if len(queue) == 0:  # Vérifier si vide
        return None
    return queue.popleft()  # O(1) - retirer de l'avant

# Front : Voir l'élément de l'avant sans le retirer
def queue_front(queue):
    """Voir l'élément de l'avant sans le retirer"""
    if len(queue) == 0:
        return None
    return queue[0]  # Retourner le premier élément

# Is empty : Vérifier si la file est vide
def queue_is_empty(queue):
    """Vérifier si la file est vide"""
    return len(queue) == 0

# Size : Obtenir le nombre d'éléments
def queue_size(queue):
    """Obtenir le nombre d'éléments dans la file"""
    return len(queue)

# Test
queue_enqueue(queue, 1)
queue_enqueue(queue, 2)
queue_enqueue(queue, 3)
print(f"File : {list(queue)}")
print(f"Avant : {queue_front(queue)}")
print(f"Retiré : {queue_dequeue(queue)}")
print(f"File après dequeue : {list(queue)}")

# Note : En pratique, vous pouvez utiliser deque directement !
# queue = deque()
# queue.append(1)    # enqueue
# queue.popleft()   # dequeue
# queue[0]          # front
````

````python
# File d'attente simple en utilisant une Liste (moins efficace - pour l'apprentissage uniquement !)
# Utiliser une liste normale pour une file est LENT car pop(0) est O(n)

# Créer une file en utilisant une liste (non recommandé pour un usage réel !)
queue_list = []  # C'est notre file (en utilisant une liste)

# Enqueue : Ajouter à la fin (O(1) - rapide !)
def simple_enqueue(queue_list, item):
    """Ajouter un élément à l'arrière de la file"""
    queue_list.append(item)  # O(1) - ajouter à la fin

# Dequeue : Retirer de l'avant (O(n) - LENT !)
def simple_dequeue(queue_list):
    """Retirer et retourner l'élément de l'avant"""
    if len(queue_list) == 0:
        return None
    return queue_list.pop(0)  # O(n) - décale tous les éléments ! Très lent !

# Test
simple_enqueue(queue_list, 1)
simple_enqueue(queue_list, 2)
print(f"File : {queue_list}")
print(f"Retiré : {simple_dequeue(queue_list)}")
print(f"File après dequeue : {queue_list}")

# ⚠️ ATTENTION : Utiliser list.pop(0) est O(n) - très lent pour de grandes files !
# Utilisez toujours deque pour les files dans le code réel !
````

## Partie 3 : Listes chaînées

### Concept de Liste chaînée

**Liste chaînée** = Chaîne de nœuds, chacun contenant des données + un pointeur vers le suivant.

**Visuel (Liste chaînée simple) :**
```
head → [1|→] → [2|→] → [3|→] → [4|→] → None
        ↑      ↑      ↑      ↑
      data  data  data  data
      next  next  next  next
```

**Structure d'un nœud :**
- Données (valeur)
- Pointeur/référence vers le nœud suivant

**Qu'est-ce qu'un "pointeur" ou une "référence" ?** 

**Analogie simple :** Pensez à un pointeur comme à une adresse postale !
- Quand vous écrivez une lettre, vous ne mettez pas toute la maison dans l'enveloppe
- Au lieu de cela, vous écrivez l'adresse (pointeur) sur l'enveloppe
- Le facteur suit l'adresse pour trouver la maison (nœud)
- De même, un pointeur ne contient pas le nœud lui-même, juste "où le trouver"

**Exemple concret :**
```python
node1 = ListNode(1)  # Créer un nœud avec la valeur 1
node2 = ListNode(2)  # Créer un nœud avec la valeur 2
node1.next = node2   # Le pointeur "next" de node1 pointe maintenant vers node2
```

**Ce qui se passe en mémoire :**
- `node1` stocke : value=1, next=adresse_de_node2
- `node2` stocke : value=2, next=None
- Quand vous accédez à `node1.next`, Python suit l'adresse pour trouver node2
- Cela crée la chaîne : node1 → node2 → None

**Pourquoi utiliser des pointeurs ?**
- Sans pointeurs : Vous devriez copier tous les nœuds (lent, gaspille la mémoire)
- Avec pointeurs : Juste stocker les adresses (rapide, efficace)
- Comme les coordonnées GPS vs transporter tout le bâtiment avec vous !

**Avantages :**
- Taille dynamique
- Insertion/suppression efficace à n'importe quelle position

**Inconvénients :**
- Pas d'accès aléatoire (pas d'indexation)
- Mémoire supplémentaire pour les pointeurs

**Types :**
- Liste chaînée simple : Chaque nœud pointe vers le suivant
- Liste chaînée double : Chaque nœud pointe vers le suivant et le précédent

````python
# Nœud de Liste chaînée en utilisant un Dictionnaire
# Chaque nœud est un dictionnaire avec les clés 'val' et 'next'
# Pas besoin de classes ! Juste des dictionnaires !

# Fonction helper pour créer un nœud (optionnel, mais rend le code plus propre)
def create_node(val, next_node=None):
    """Créer un dictionnaire de nœud"""
    return {'val': val, 'next': next_node}

# Créer une liste chaînée simple : 1 -> 2 -> 3
# Étape 1 : Créer trois nœuds séparés (comme dictionnaires)
node1 = {'val': 1, 'next': None}  # Nœud avec valeur 1, next=None
node2 = {'val': 2, 'next': None}  # Nœud avec valeur 2, next=None
node3 = {'val': 3, 'next': None}  # Nœud avec valeur 3, next=None

# Étape 2 : Les connecter en utilisant des pointeurs
# node1['next'] = node2 signifie : "Le pointeur de node1 pointe maintenant vers node2"
# Cela crée le lien : node1 → node2
node1['next'] = node2  # Lier node1 à node2

# node2['next'] = node3 signifie : "Le pointeur de node2 pointe maintenant vers node3"
# Cela crée le lien : node2 → node3
node2['next'] = node3  # Lier node2 à node3

# Résultat : node1 → node2 → node3 → None
# node3['next'] est toujours None, ce qui signifie "fin de la liste"

# Parcourir et afficher
# Parcourir signifie : visiter chaque nœud un par un, en suivant les pointeurs
def print_linked_list(head):
    """
    Afficher toutes les valeurs de la liste chaînée en suivant les pointeurs.
    
    Comment ça fonctionne :
    1. Commencer à head (premier nœud)
    2. Afficher la valeur du nœud actuel
    3. Passer au nœud suivant en utilisant le pointeur (current = current['next'])
    4. Répéter jusqu'à atteindre None (fin de la liste)
    """
    current = head  # Commencer au premier nœud (head)
    
    # Continuer jusqu'à atteindre None (fin de la liste)
    while current:
        # Afficher la valeur avec une flèche s'il y a un nœud suivant
        print(current['val'], end=" -> " if current['next'] else "\n")
        
        # Passer au nœud suivant en suivant le pointeur
        # current['next'] nous donne le dictionnaire du nœud suivant
        # On l'assigne à current pour "avancer"
        current = current['next']  # Suivre le pointeur vers le nœud suivant

print_linked_list(node1)  # Sortie : 1 -> 2 -> 3

# Alternative : Utiliser la fonction helper (plus propre !)
node1 = create_node(1)
node2 = create_node(2)
node3 = create_node(3)
node1['next'] = node2
node2['next'] = node3
print_linked_list(node1)  # Sortie : 1 -> 2 -> 3
````

````python
# Opérations sur les Listes chaînées (en utilisant des dictionnaires)

# Fonction helper pour créer un nœud
def create_node(val, next_node=None):
    """Créer un dictionnaire de nœud"""
    return {'val': val, 'next': next_node}

# 1. Parcours - O(n) - Visiter chaque nœud de head à la fin
def traverse(head):
    """
    Visiter chaque nœud de la liste chaînée.
    
    Temps : O(n) - doit visiter chaque nœud une fois
    Espace : O(1) - n'utilise qu'une variable (current)
    
    Comment les pointeurs fonctionnent ici :
    - current commence à head (premier nœud)
    - current['next'] nous donne le dictionnaire du nœud suivant
    - On suit la chaîne de pointeurs jusqu'à atteindre None
    """
    current = head  # Commencer à head (premier nœud)
    
    # Continuer jusqu'à atteindre None (fin de la liste)
    while current:
        print(current['val'])  # Accéder à la valeur en utilisant la clé du dictionnaire
        # Passer au nœud suivant en suivant le pointeur
        # C'est la CLÉ : current['next'] est un pointeur, on le suit !
        current = current['next']  # Suivre le pointeur vers le nœud suivant

# 2. Recherche - O(n) - Chercher une valeur dans la liste
def search(head, target):
    """
    Rechercher une valeur dans la liste chaînée.
    
    Temps : O(n) cas le pire - pourrait devoir vérifier tous les nœuds
    Espace : O(1) - n'utilise qu'une variable
    
    Pourquoi O(n) ? Contrairement aux tableaux, on ne peut pas sauter au milieu. Doit vérifier un par un !
    """
    current = head  # Commencer au premier nœud
    
    while current:
        if current['val'] == target:  # Trouvé ! Accéder à la valeur en utilisant la clé du dictionnaire
            return True
        
        # Pas encore trouvé - suivre le pointeur vers le nœud suivant
        current = current['next']  # Avancer en utilisant le pointeur
    
    return False  # Atteint la fin, pas trouvé

# 3. Insérer au début - O(1) - Ajouter un nouveau nœud comme head
def insert_at_head(head, val):
    """
    Insérer un nouveau nœud au début de la liste.
    
    Temps : O(1) - temps constant ! Juste créer le nœud et mettre à jour le pointeur
    Espace : O(1) - ne crée qu'un nouveau nœud
    
    Visuel :
    Avant : head → [1|→] → [2|→] → None
    Après :  head → [0|→] → [1|→] → [2|→] → None
    
    Étapes :
    1. Créer un nouveau nœud avec la valeur
    2. Pointer le nouveau nœud vers l'ancien head (new_node['next'] = head)
    3. Retourner le nouveau nœud comme nouveau head
    """
    new_node = create_node(val)  # Créer un dictionnaire de nouveau nœud
    
    # Pointer le pointeur du nouveau nœud vers l'ancien head
    # Cela connecte : new_node → old_head
    new_node['next'] = head
    
    # Retourner le nouveau nœud comme nouveau head
    # Maintenant new_node est le premier nœud de la liste !
    return new_node

# 4. Insérer après un nœud - O(1) - Ajouter un nouveau nœud après le nœud donné
def insert_after(node, val):
    """
    Insérer un nouveau nœud après un nœud spécifique.
    
    Temps : O(1) - temps constant ! Juste mettre à jour les pointeurs
    Espace : O(1) - ne crée qu'un nouveau nœud
    
    Visuel :
    Avant : node → [A|→] → [B|→] → None
    Après :  node → [A|→] → [new|→] → [B|→] → None
    
    Étapes :
    1. Créer un nouveau nœud
    2. Sauvegarder ce vers quoi node pointe actuellement (node['next'])
    3. Pointer le nouveau nœud vers ce nœud sauvegardé
    4. Pointer le nœud original vers le nouveau nœud
    """
    new_node = create_node(val)  # Créer un dictionnaire de nouveau nœud
    
    # Sauvegarder ce vers quoi le nœud original pointe
    # On en a besoin car on insère ENTRE node et node['next']
    new_node['next'] = node['next']  # Le nouveau nœud pointe vers ce vers quoi l'ancien nœud pointait
    
    # Maintenant pointer le nœud original vers le nouveau nœud
    # Cela crée : node → new_node → (ce vers quoi node['next'] pointait avant)
    node['next'] = new_node

# 5. Supprimer un nœud (nœud donné à supprimer) - O(1) si on a le nœud
def delete_node(node):
    """
    Supprimer un nœud en copiant la valeur du nœud suivant et en le sautant.
    
    Temps : O(1) - temps constant ! Juste copier la valeur et mettre à jour le pointeur
    Espace : O(1) - pas d'espace supplémentaire nécessaire
    
    Visuel :
    Avant : ... → [A|→] → [B|→] → [C|→] → ...
    Après :  ... → [B|→] → [C|→] → ...
    
    Astuce : Au lieu de supprimer le nœud actuel (difficile !), on :
    1. Copie la valeur du nœud suivant vers le nœud actuel
    2. Saute le nœud suivant en pointant current['next'] vers next['next']
    
    Note : Cela ne fonctionne pas si le nœud est le dernier nœud (pas de nœud suivant à copier)
    """
    if node['next']:  # S'assurer qu'il y a un nœud suivant
        # Copier la valeur du nœud suivant vers le nœud actuel
        node['val'] = node['next']['val']
        
        # Sauter le nœud suivant en pointant current['next'] vers next['next']
        # Cela "supprime" le nœud suivant en retirant le pointeur vers lui
        node['next'] = node['next']['next']

# Test
# Créer une liste chaînée : 1 -> 2 -> 3
head = create_node(1)
head['next'] = create_node(2)
head['next']['next'] = create_node(3)

print("Original :")
print_linked_list(head)

head = insert_at_head(head, 0)
print("Après insertion de 0 au début :")
print_linked_list(head)

insert_after(head['next'], 1.5)  # Insérer après le premier nœud
print("Après insertion de 1.5 après le premier nœud :")
print_linked_list(head)
````

````python
# Modèles courants de Listes chaînées

# Modèle 1 : Deux Pointeurs (Rapide & Lent)
# Trouver le milieu de la liste chaînée
def find_middle(head):
    """Trouver le nœud du milieu en utilisant deux pointeurs"""
    slow = fast = head
    while fast and fast['next']:  # Accéder à 'next' en utilisant la clé du dictionnaire
        slow = slow['next']
        fast = fast['next']['next']
    return slow

# Modèle 2 : Inverser une Liste chaînée
# Visualisation étape par étape :

# Initial : 1 → 2 → 3 → 4 → None
#          ↑
#        head, current
#        prev = None

# Étape 1 : Sauvegarder next, inverser le lien, déplacer les pointeurs
#  1 → 2 → 3 → 4 → None
#  ↑   ↑
# prev current
#  None ← 1  2 → 3 → 4 (lien inversé !)

# Étape 2 : Continuer...
#  None ← 1 ← 2  3 → 4
#         ↑   ↑
#        prev current

# Étape 3 : Continuer...
#  None ← 1 ← 2 ← 3  4 → None
#              ↑   ↑
#            prev current

# Étape 4 : Étape finale
#  None ← 1 ← 2 ← 3 ← 4
#                  ↑   ↑
#                prev current (None)
# Retourner prev (nouveau head = 4)

def reverse_linked_list(head):
    """Inverser une liste chaînée"""
    prev = None
    current = head
    
    while current:
        next_node = current['next']  # Sauvegarder next (avant de le perdre !) - utiliser la clé du dictionnaire
        current['next'] = prev        # Inverser le lien
        prev = current                 # Déplacer prev vers l'avant
        current = next_node            # Déplacer current vers l'avant
    
    return prev  # Nouveau head

# Test
head = create_node(1)
head['next'] = create_node(2)
head['next']['next'] = create_node(3)
head['next']['next']['next'] = create_node(4)

print("Original :")
print_linked_list(head)

middle = find_middle(head)
print(f"Milieu : {middle['val']}")  # Accéder à la valeur en utilisant la clé du dictionnaire

reversed_head = reverse_linked_list(head)
print("Inversé :")
print_linked_list(reversed_head)
````

````python
# Modèle 3 : Détecter un Cycle (Détection de Cycle de Floyd)
# Qu'est-ce qu'un cycle ? Quand un nœud pointe vers un nœud précédent, créant une boucle.
# Exemple : 1 → 2 → 3 → 2 → 3 → 2... (3 pointe vers 2, créant une boucle infinie)
# Liste normale : 1 → 2 → 3 → None (se termine à None, pas de cycle)
# Pourquoi détecter les cycles ? Ils causent des boucles infinies et des plantages !

def has_cycle(head):
    """
    Utiliser deux pointeurs : slow se déplace de 1 pas, fast se déplace de 2 pas
    S'il y a un cycle, ils se rencontreront (comme des coureurs sur une piste circulaire)
    """
    if not head or not head['next']:  # Accéder à 'next' en utilisant la clé du dictionnaire
        return False
    
    slow = head
    fast = head['next']
    
    while fast and fast['next']:
        if slow == fast:  # Les pointeurs se rencontrent - cycle détecté !
            return True
        slow = slow['next']
        fast = fast['next']['next']
    
    return False

# Test avec cycle
node1 = create_node(1)
node2 = create_node(2)
node3 = create_node(3)
node1['next'] = node2
node2['next'] = node3
node3['next'] = node2  # Cycle ! node3 pointe vers node2

print(f"A un cycle : {has_cycle(node1)}")

# Test sans cycle
node1 = create_node(1)
node2 = create_node(2)
node3 = create_node(3)
node1['next'] = node2
node2['next'] = node3
# node3['next'] est None - pas de cycle

print(f"A un cycle : {has_cycle(node1)}")
````

from collections import deque

def my_stack_push(stack_queue, x):
    """Empiler x sur la pile en utilisant une file"""
    stack_queue.append(x)
    # Faire tourner pour rendre le dernier élément premier
    for _ in range(len(stack_queue) - 1):
        stack_queue.append(stack_queue.popleft())

def my_stack_pop(stack_queue):
    """Dépiler de la pile en utilisant une file"""
    return stack_queue.popleft()

def my_stack_top(stack_queue):
    """Obtenir l'élément du sommet"""
    return stack_queue[0]

def my_stack_empty(stack_queue):
    """Vérifier si la pile est vide"""
    return len(stack_queue) == 0

### Solution 2 : Supprimer le Nième Nœud
```python
def remove_nth_from_end(head, n):
    """Supprimer le nième nœud depuis la fin en utilisant des dictionnaires"""
    dummy = create_node(0)
    dummy['next'] = head
    
    first = second = dummy
    
    # Déplacer le premier pointeur de n+1 pas en avant
    for _ in range(n + 1):
        first = first['next']
    
    # Déplacer les deux jusqu'à ce que first atteigne la fin
    while first:
        first = first['next']
        second = second['next']
    
    # Supprimer le nœud
    second['next'] = second['next']['next']
    
    return dummy['next']
```

### Solution 3 : Fusionner Deux Listes Triées
```python
def merge_two_lists(list1, list2):
    """Fusionner deux listes triées en utilisant des dictionnaires"""
    dummy = create_node(0)
    current = dummy
    
    while list1 and list2:
        if list1['val'] <= list2['val']:
            current['next'] = list1
            list1 = list1['next']
        else:
            current['next'] = list2
            list2 = list2['next']
        current = current['next']
    
    # Attacher le reste
    current['next'] = list1 if list1 else list2
    
    return dummy['next']
```

## Points clés à retenir

✅ **Piles :**
- LIFO (Dernier Entré Premier Sorti)
- Utiliser pour la correspondance, les opérations d'annulation
- O(1) push/pop

✅ **Files d'attente :**
- FIFO (Premier Entré Premier Sorti)
- Utiliser pour BFS, la planification
- Utiliser `deque` pour les opérations O(1)

✅ **Listes chaînées :**
- Taille dynamique, insertion/suppression efficace
- Pas d'accès aléatoire
- Le modèle de deux pointeurs est puissant
- Attention aux cas limites : liste vide, nœud unique
