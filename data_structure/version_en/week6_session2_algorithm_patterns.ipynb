## Part 1: Two Pointers Technique

### Two Pointers Pattern

**Two Pointers** = Use two pointers moving through data structure.

**Pattern 1: Opposite Ends**
```
Array: [2, 7, 11, 15], target = 9
       ↑              ↑
      left          right
      (2)           (15)
      2+15=17 > 9, move right ←
      2+11=13 > 9, move right ←
      2+7=9 ✓ Found!
```

**Pattern 2: Fast & Slow (Tortoise & Hare)**

**The Idea:** Move two pointers at different speeds. When fast reaches the end, slow is at the middle!

**Step-by-Step Walkthrough:**
```
Initial: 1 → 2 → 3 → 4 → 5
         ↑
    slow,fast (both start at node 1)

Step 1: Move slow +1, fast +2
        1 → 2 → 3 → 4 → 5
            ↑   ↑
          slow fast

Step 2: Move slow +1, fast +2
        1 → 2 → 3 → 4 → 5
                ↑       ↑
              slow     fast

Step 3: Fast can't move (reached end), stop!
        1 → 2 → 3 → 4 → 5
                ↑
              slow = middle! ✓
```

**Why This Works:**
- Fast moves 2x speed of slow (like a race car vs bicycle!)
- When fast travels full length (n nodes), slow travels half (n/2 nodes)
- Therefore, slow ends at the middle!

**Concrete Example:**
- List with 5 nodes: 1 → 2 → 3 → 4 → 5
- Fast moves: 1 → 3 → 5 (2 steps each time, reaches end)
- Slow moves: 1 → 2 → 3 (1 step each time, ends at middle!)
- Result: Slow is at node 3, which is the middle ✓

**Edge Cases:**
- Even length: `1 → 2 → 3 → 4` → slow ends at node 2 (first middle)
- Single node: `1` → slow = fast = node 1

**Use Cases:**
- Sorted array problems
- Palindrome checking
- Finding pairs
- Cycle detection

````python
# Pattern 1: Opposite Ends (Sorted Array)
# Two Sum in sorted array

def two_sum_sorted(arr, target):
    """
    Find two numbers that sum to target
    Array is sorted
    """
    left, right = 0, len(arr) - 1
    
    while left < right:
        current_sum = arr[left] + arr[right]
        
        if current_sum == target:
            return [left, right]
        elif current_sum < target:
            left += 1  # Need larger sum
        else:
            right -= 1  # Need smaller sum
    
    return []

# Test
arr = [2, 7, 11, 15]
print(f"Two sum indices: {two_sum_sorted(arr, 9)}")
````

````python
# Pattern 2: Fast & Slow (Linked List)
# Find middle of linked list

# Helper function to create a node (using dictionary)
def create_node(val, next_node=None):
    """Create a linked list node dictionary"""
    return {'val': val, 'next': next_node}

def find_middle(head):
    """Find middle node using two pointers"""
    slow = fast = head
    
    while fast and fast['next']:  # Access 'next' using dictionary key
        slow = slow['next']      # Move 1 step - access 'next' using dictionary key
        fast = fast['next']['next'] # Move 2 steps - access 'next' using dictionary key
    
    return slow

# Pattern 3: Remove duplicates from sorted array
def remove_duplicates(arr):
    """
    Remove duplicates in-place, return new length
    """
    if not arr:
        return 0
    
    write_ptr = 1  # Position to write next unique element
    
    for read_ptr in range(1, len(arr)):
        if arr[read_ptr] != arr[read_ptr - 1]:
            arr[write_ptr] = arr[read_ptr]
            write_ptr += 1
    
    return write_ptr

# Test
arr = [1, 1, 2, 2, 3, 4, 4, 5]
new_len = remove_duplicates(arr)
print(f"New length: {new_len}")
print(f"Array: {arr[:new_len]}")

arr[:5]
````

## Part 2: Sliding Window Pattern

### Sliding Window Concept

**Sliding Window** = Maintain a window that slides through array/string.

**Real-World Analogy:** Like a security guard watching a security camera feed!
- The camera shows a fixed-size window of the scene
- As the scene moves, instead of resetting the camera, you just slide the view
- You keep track of what's in the current view without recalculating everything

**The Idea:** Instead of recalculating everything, slide the window and update incrementally!

**Why is this faster?** 
- **Naive approach:** For each window, add all k numbers from scratch
  - Window 1: 2+1+5 = 8 (3 additions)
  - Window 2: 1+5+1 = 7 (3 additions again!)
  - Window 3: 5+1+3 = 9 (3 additions again!)
  - Window 4: 1+3+2 = 6 (3 additions again!)
  - Total: 3 additions × 4 windows = 12 operations ✗
  
- **Sliding window:** Calculate once, then reuse!
  - Window 1: 2+1+5 = 8 (calculate once)
  - Window 2: 8 - 2 + 1 = 7 (just subtract left, add right!)
  - Window 3: 7 - 1 + 3 = 9 (just subtract left, add right!)
  - Window 4: 9 - 5 + 2 = 6 (just subtract left, add right!)
  - Total: 3 + 3 = 6 operations ✓ (50% faster!)

**Key Insight:** Don't recalculate! Reuse the previous sum by subtracting what leaves and adding what enters!

**Visual (Fixed Size k=3):**
```
Array: [2, 1, 5, 1, 3, 2]
       [---]              Window 1: sum = 2+1+5 = 8
          [---]           Window 2: sum = 8-2+1 = 7 
                         (subtract 2 that left, add 1 that entered)
             [---]        Window 3: sum = 7-1+3 = 9
                         (subtract 1 that left, add 3 that entered)
                [---]      Window 4: sum = 9-5+2 = 6
                         (subtract 5 that left, add 2 that entered)
Max sum = 9

Think of it like a moving train:
- Window 1: [2,1,5] → sum = 8
- Window 2: [1,5,1] → remove 2 (left), add 1 (entered) → 8-2+1 = 7
- Window 3: [5,1,3] → remove 1 (left), add 3 (entered) → 7-1+3 = 9
```
**Key Insight:** Don't recalculate! Subtract what leaves, add what enters. O(n) instead of O(nk)!

**Visual (Variable Size - Expand & Shrink):**
```
String: "abcabcbb"
        [a]            Window: {a}, len=1
        [ab]           Expand: {a,b}, len=2
        [abc]          Expand: {a,b,c}, len=3 ✓
        [abca]         Expand: duplicate 'a' found!
         [bca]         Shrink left until 'a' removed, len=3 ✓
         [bcab]        Expand: duplicate 'b' found!
          [cab]        Shrink left until 'b' removed, len=3 ✓
Max length = 3
```

**Step-by-Step (Variable Window):**
1. **Expand:** Move `right` pointer, add to window
2. **Check:** Is window valid? (no duplicates, sum ≤ target, etc.)
3. **Shrink:** If invalid, move `left` pointer until valid again
4. **Update:** Track best answer (max length, min sum, etc.)

**Types:**
1. **Fixed Size**: Window size is constant (easier!)
2. **Variable Size**: Expand/shrink based on condition (more flexible)

**Use Cases:**
- Substring/subarray problems
- Maximum/minimum in window
- Longest/shortest subarray with condition

````python
# Pattern 1: Fixed Size Window
# Maximum sum of subarray of size k

def max_sum_subarray(arr, k):
    """
    Find maximum sum of subarray of size k
    """
    if len(arr) < k:
        return 0
    
    # Calculate sum of first window
    window_sum = sum(arr[:k])
    max_sum = window_sum
    
    # Slide the window
    for i in range(k, len(arr)):
        window_sum = window_sum - arr[i - k] + arr[i]
        max_sum = max(max_sum, window_sum)
    
    return max_sum

# Test
arr = [2, 1, 5, 1, 3, 2]
k = 3
print(f"Max sum of subarray size {k}: {max_sum_subarray(arr, k)}")
````

````python
# Pattern 2: Variable Size Window
# Longest substring without repeating characters

def length_of_longest_substring(s):
    """
    Find length of longest substring without repeating characters
    """
    char_set = set()
    left = 0
    max_len = 0
    
    for right in range(len(s)):
        # Shrink window until no duplicates
        while s[right] in char_set:
            char_set.remove(s[left])
            left += 1
        
        # Expand window
        char_set.add(s[right])
        max_len = max(max_len, right - left + 1)
    
    return max_len

# Test
s = "abcabcbb"
print(f"Longest substring length: {length_of_longest_substring(s)}")
````

````python
# Pattern 3: Minimum Window Substring
# Find minimum window in s containing all characters in t

def min_window(s, t):
    """
    Find minimum window substring containing all chars in t
    """
    if not s or not t:
        return ""
    
    # Count characters in t
    need = {}
    for char in t:
        need[char] = need.get(char, 0) + 1
    
    have = {}
    left = 0
    min_len = float('inf')
    min_window_start = 0
    required = len(need)
    formed = 0
    
    for right in range(len(s)):
        char = s[right]
        have[char] = have.get(char, 0) + 1
        
        # Check if current char satisfies requirement
        if char in need and have[char] == need[char]:
            formed += 1
        
        # Try to shrink window
        while left <= right and formed == required:
            # Update minimum window
            if right - left + 1 < min_len:
                min_len = right - left + 1
                min_window_start = left
            
            # Remove leftmost character
            left_char = s[left]
            have[left_char] -= 1
            if left_char in need and have[left_char] < need[left_char]:
                formed -= 1
            
            left += 1
    
    return "" if min_len == float('inf') else s[min_window_start:min_window_start + min_len]

# Test
s = "ADOBECODEBANC"
t = "ABC"
print(f"Min window: '{min_window(s, t)}'")
````

## Part 3: BFS/DFS for Graphs

### When to Use BFS vs DFS

**Visual Comparison:**
```
Graph:     0
          / \
         1   2
        / \   \
       3   4   5

BFS Order: 0 → 1 → 2 → 3 → 4 → 5  (level by level)
DFS Order: 0 → 1 → 3 → 4 → 2 → 5  (go deep first)
```

**BFS (Breadth-First Search):**
- **How:** Use queue, explore level by level
- **When:** Shortest path (unweighted graphs), level-order traversal
- **Think:** "Visit all neighbors before going deeper"

**DFS (Depth-First Search):**
- **How:** Use recursion/stack, go deep before backtracking
- **When:** Exploring all paths, topological sorting, detecting cycles
- **Think:** "Go as deep as possible, then backtrack"

**Quick Decision:**
- Need shortest path? → BFS
- Need all paths or deep exploration? → DFS

````python
from collections import deque

# BFS: Shortest Path (Unweighted Graph)
def shortest_path(graph, start, end):
    """
    Find shortest path using BFS
    """
    if start == end:
        return [start]
    
    queue = deque([(start, [start])])
    visited = {start}
    
    while queue:
        node, path = queue.popleft()
        
        for neighbor in graph[node]:
            if neighbor == end:
                return path + [neighbor]
            
            if neighbor not in visited:
                visited.add(neighbor)
                queue.append((neighbor, path + [neighbor]))
    
    return []  # No path found

# Test
graph = {
    0: [1, 2],
    1: [0, 3],
    2: [0, 3],
    3: [1, 2, 4],
    4: [3]
}

print(f"Shortest path 0->4: {shortest_path(graph, 0, 4)}")
````

````python
# DFS: All Paths Between Two Nodes
def all_paths(graph, start, end):
    """
    Find all paths using DFS
    """
    paths = []
    
    def dfs(node, path, visited):
        if node == end:
            paths.append(path[:])
            return
        
        for neighbor in graph[node]:
            if neighbor not in visited:
                visited.add(neighbor)
                path.append(neighbor)
                dfs(neighbor, path, visited)
                path.pop()
                visited.remove(neighbor)
    
    dfs(start, [start], {start})
    return paths

print(f"All paths 0->3: {all_paths(graph, 0, 3)}")
````

## Part 4: Problem-Solving Framework

### Step-by-Step Approach

1. **Understand the Problem**
   - Read constraints carefully
   - Identify input/output
   - Ask clarifying questions

2. **Choose Data Structure**
   - Need fast lookup? → Hash map
   - Need order? → Array/List
   - Need priority? → Heap
   - Need relationships? → Graph

3. **Choose Algorithm**
   - Sorted array? → Two pointers or binary search
   - Substring/subarray? → Sliding window
   - Shortest path? → BFS
   - All paths? → DFS

4. **Handle Edge Cases**
   - Empty input
   - Single element
   - All same elements
   - Very large input

5. **Write Clean Code**
   - Meaningful variable names
   - Comments for complex logic
   - Test with examples

````python
# Example: Applying the Framework
# Problem: Find all anagrams in a string

# Reminder: Anagrams are words with same letters in different order.
# Example: "abc" and "bac" are anagrams (both have a, b, c)
# In string "cbaebabacd", find all substrings that are anagrams of "abc"
# Answer: "cba" (index 0) and "bac" (index 6) are anagrams of "abc"
# ["cba", "bac"]
def find_anagrams(s, p):
    """
    Find all start indices of anagrams of p in s
    
    Step 1: Understand
    - Input: string s, string p
    - Output: list of start indices
    - Constraint: p.length <= s.length
    
    Step 2: Choose DS
    - Need frequency count → Hash map
    
    Step 3: Choose Algorithm
    - Fixed window of size len(p) → Sliding window
    
    Step 4: Edge Cases
    - len(p) > len(s) → return []
    - Empty strings
    """
    if len(p) > len(s):
        return []
    
    # Frequency of p
    p_freq = {}
    for char in p:
        p_freq[char] = p_freq.get(char, 0) + 1
    
    # Sliding window
    window_freq = {}
    result = []
    
    # Initialize window
    for i in range(len(p)):
        window_freq[s[i]] = window_freq.get(s[i], 0) + 1
    
    # Check first window
    if window_freq == p_freq:
        result.append(0)
    
    # Slide window
    for i in range(len(p), len(s)):
        # Remove leftmost
        left_char = s[i - len(p)]
        window_freq[left_char] -= 1
        if window_freq[left_char] == 0:
            del window_freq[left_char]
        
        # Add rightmost
        right_char = s[i]
        window_freq[right_char] = window_freq.get(right_char, 0) + 1
        
        # Check if anagram
        if window_freq == p_freq:
            result.append(i - len(p) + 1)
    
    return result

# Test
s = "cbaebabacd"
p = "abc"
print(f"Anagram indices: {find_anagrams(s, p)}")
````

```python
def character_replacement(s, k):
    freq = {}
    left = 0
    max_len = 0
    max_freq = 0
    
    for right in range(len(s)):
        freq[s[right]] = freq.get(s[right], 0) + 1
        max_freq = max(max_freq, freq[s[right]])

        if (right - left + 1) - max_freq > k:
            freq[s[left]] -= 1
            left += 1
```

### Solution 3: Word Ladder
```python
from collections import deque

def ladder_length(begin_word, end_word, word_list):
    word_set = set(word_list)
    if end_word not in word_set:
        return 0
    
    queue = deque([(begin_word, 1)])
    visited = {begin_word}
    
    while queue:
        word, length = queue.popleft()
        
        if word == end_word:
            return length
        
        # Try all one-letter changes
        for i in range(len(word)):
            for c in 'abcdefghijklmnopqrstuvwxyz':
                new_word = word[:i] + c + word[i+1:]
                if new_word in word_set and new_word not in visited:
                    visited.add(new_word)
                    queue.append((new_word, length + 1))
    
    return 0
```

- Opposite ends: sorted arrays, pairs

- Great for substring/subarray problems

✅ **BFS vs DFS:**
- BFS: Shortest path, level-order
