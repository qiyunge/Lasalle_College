## Part 1: Stacks (LIFO - Last In First Out)

### Stack Concept

**Stack** = LIFO (Last In First Out) - like a stack of plates!

**Visual:**
```
push(3) → push(2) → push(1)
         ┌─────┐
         │  1  │ ← Top (last added)
         ├─────┤
         │  2  │
         ├─────┤
         │  3  │ ← Bottom (first added)
         └─────┘
pop() removes 1 (top), peek() returns 1
```

**Operations:**
- `push(item)` - Add to top (O(1))
- `pop()` - Remove from top (O(1))
- `peek()` / `top()` - View top without removing (O(1))
- `is_empty()` - Check if empty (O(1))

**Use Cases:**
- Matching parentheses/brackets
- Undo operations
- Function call stack
- Expression evaluation

````python
# Stack Implementation using List
# A stack is just a list! We use list operations to simulate stack behavior.

# Stack operations (using a list directly):
# - push: use append() to add to end (top)
# - pop: use pop() to remove from end (top)
# - peek: use [-1] to view last element
# - is_empty: check if list is empty
# - size: use len()

# Create a stack (just an empty list!)
from typing import List


stack = []  # This is our stack!

# Push: Add item to top of stack
def stack_push(stack: List, item):
    """Add item to top of stack"""
    stack.append(item)  # O(1) - add to end

# Pop: Remove and return top item
def stack_pop(stack):
    """Remove and return top item"""
    if len(stack) == 0:  # Check if empty
        return None
    return stack.pop()  # O(1) - remove from end

# Peek: View top item without removing
def stack_peek(stack):
    """View top item without removing"""
    if len(stack) == 0:
        return None
    return stack[-1]  # Return last element

# Is empty: Check if stack is empty
def stack_is_empty(stack):
    """Check if stack is empty"""
    return len(stack) == 0

# Size: Get number of items
def stack_size(stack):
    """Get number of items in stack"""
    return len(stack)

# Test
stack_push(stack, 1)
stack_push(stack, 2)
stack_push(stack, 3)
print(f"Stack: {stack}")
print(f"Top: {stack_peek(stack)}")
print(f"Popped: {stack_pop(stack)}")
print(f"Stack after pop: {stack}")

# Note: In practice, you can use the list directly!
# stack = []
# stack.append(1)  # push
# stack.pop()     # pop
# stack[-1]       # peek
````

````python
# Problem: Valid Parentheses
# Check if parentheses are balanced: "()[]{}"

def is_valid_parentheses(s):
    stack = []
    pairs = {')': '(', '}': '{', ']': '['}
    
    for char in s:
        if char in pairs:  # Closing bracket
            if not stack or stack.pop() != pairs[char]:  
                return False
        else:  # Opening bracket
            stack.append(char)
    
    return len(stack) == 0

# Test
print(f"'(A#' valid: {is_valid_parentheses('(A#')}")
print(f"'()[]{{}}' valid: {is_valid_parentheses('()[]{}')}")
print(f"'(]' valid: {is_valid_parentheses('(]')}")
print(f"'([)]' valid: {is_valid_parentheses('([)]')}")
````

## Part 2: Queues (FIFO - First In First Out)

### Queue Concept

**Queue** = FIFO (First In First Out) - like a line at a store!

**Visual:**
```
enqueue(1) → enqueue(2) → enqueue(3)
Front → [1] [2] [3] ← Back
         ↑           ↑
      dequeue()   enqueue()
      removes 1   adds here
```

**Operations:**
- `enqueue(item)` - Add to back (O(1))
- `dequeue()` - Remove from front (O(1) with collections.deque, O(n) with list)
- `front()` / `peek()` - View front without removing (O(1))
- `is_empty()` - Check if empty (O(1))

**Use Cases:**
- BFS (Breadth-First Search)
- Task scheduling
- Print queue
- Level-order tree traversal

````python
# Queue Implementation using collections.deque (efficient!)
from collections import deque

# A queue is a deque! We use deque operations directly.
# deque = "double-ended queue" - efficient for both ends

# Queue operations (using deque directly):
# - enqueue: use append() to add to back
# - dequeue: use popleft() to remove from front
# - front: use [0] to view first element
# - is_empty: check if deque is empty
# - size: use len()

# Create a queue (just a deque!)
queue = deque()  # This is our queue!

# Enqueue: Add item to back of queue
def queue_enqueue(queue, item):
    """Add item to back of queue"""
    queue.append(item)  # O(1) - add to end

# Dequeue: Remove and return front item
def queue_dequeue(queue):
    """Remove and return front item"""
    if len(queue) == 0:  # Check if empty
        return None
    return queue.popleft()  # O(1) - remove from front

# Front: View front item without removing
def queue_front(queue):
    """View front item without removing"""
    if len(queue) == 0:
        return None
    return queue[0]  # Return first element

# Is empty: Check if queue is empty
def queue_is_empty(queue):
    """Check if queue is empty"""
    return len(queue) == 0

# Size: Get number of items
def queue_size(queue):
    """Get number of items in queue"""
    return len(queue)

# Test
queue_enqueue(queue, 1)
queue_enqueue(queue, 2)
queue_enqueue(queue, 3)
print(f"Queue: {list(queue)}")
print(f"Front: {queue_front(queue)}")
print(f"Dequeued: {queue_dequeue(queue)}")
print(f"Queue after dequeue: {list(queue)}")

# Note: In practice, you can use deque directly!
# queue = deque()
# queue.append(1)    # enqueue
# queue.popleft()   # dequeue
# queue[0]          # front
````

````python
# Simple Queue using List (less efficient - for learning only!)
# Using a regular list for queue is SLOW because pop(0) is O(n)

# Create a queue using list (not recommended for real use!)
queue_list = []  # This is our queue (using list)

# Enqueue: Add to end (O(1) - fast!)
def simple_enqueue(queue_list, item):
    """Add item to back of queue"""
    queue_list.append(item)  # O(1) - add to end

# Dequeue: Remove from front (O(n) - SLOW!)
def simple_dequeue(queue_list):
    """Remove and return front item"""
    if len(queue_list) == 0:
        return None
    return queue_list.pop(0)  # O(n) - shifts all elements! Very slow!

# Test
simple_enqueue(queue_list, 1)
simple_enqueue(queue_list, 2)
print(f"Queue: {queue_list}")
print(f"Dequeued: {simple_dequeue(queue_list)}")
print(f"Queue after dequeue: {queue_list}")

# ⚠️ WARNING: Using list.pop(0) is O(n) - very slow for large queues!
# Always use deque for queues in real code!
````

## Part 3: Linked Lists

### Linked List Concept

**Linked List** = Chain of nodes, each containing data + pointer to next.

**Visual (Singly Linked List):**
```
head → [1|→] → [2|→] → [3|→] → [4|→] → None
        ↑      ↑      ↑      ↑
      data  data  data  data
      next  next  next  next
```

**Node Structure:**
- Data (value)
- Pointer/reference to next node

**What is a "pointer" or "reference"?** 

**Simple Analogy:** Think of a pointer like a street address!
- When you write a letter, you don't put the entire house in the envelope
- Instead, you write the address (pointer) on the envelope
- The mail carrier follows the address to find the house (node)
- Similarly, a pointer doesn't contain the node itself, just "where to find it"

**Concrete Example:**
```python
node1 = ListNode(1)  # Create node with value 1
node2 = ListNode(2)  # Create node with value 2
node1.next = node2   # node1's "next" pointer now points to node2
```

**What happens in memory:**
- `node1` stores: value=1, next=address_of_node2
- `node2` stores: value=2, next=None
- When you access `node1.next`, Python follows the address to find node2
- This creates the chain: node1 → node2 → None

**Why use pointers?**
- Without pointers: You'd have to copy entire nodes (slow, wastes memory)
- With pointers: Just store addresses (fast, efficient)
- Like GPS coordinates vs carrying the entire building with you!

**Advantages:**
- Dynamic size
- Efficient insertion/deletion at any position

**Disadvantages:**
- No random access (no indexing)
- Extra memory for pointers

**Types:**
- Singly Linked List: Each node points to next
- Doubly Linked List: Each node points to next and previous

````python
# Linked List Node using Dictionary
# Each node is a dictionary with 'val' and 'next' keys
# No classes needed! Just dictionaries!

# Helper function to create a node (optional, but makes code cleaner)
def create_node(val, next_node=None):
    """Create a node dictionary"""
    return {'val': val, 'next': next_node}

# Create a simple linked list: 1 -> 2 -> 3
# Step 1: Create three separate nodes (as dictionaries)
node1 = {'val': 1, 'next': None}  # Node with value 1, next=None
node2 = {'val': 2, 'next': None}  # Node with value 2, next=None
node3 = {'val': 3, 'next': None}  # Node with value 3, next=None

# Step 2: Connect them using pointers
# node1['next'] = node2 means: "node1's pointer now points to node2"
# This creates the link: node1 → node2
node1['next'] = node2  # Link node1 to node2

# node2['next'] = node3 means: "node2's pointer now points to node3"
# This creates the link: node2 → node3
node2['next'] = node3  # Link node2 to node3

# Result: node1 → node2 → node3 → None
# node3['next'] is still None, which means "end of list"

# Traverse and print
# Traversal means: visit each node one by one, following the pointers
def print_linked_list(head):
    """
    Print all values in linked list by following pointers.
    
    How it works:
    1. Start at head (first node)
    2. Print current node's value
    3. Move to next node using pointer (current = current['next'])
    4. Repeat until we reach None (end of list)
    """
    current = head  # Start at the first node (head)
    
    # Keep going until we reach None (end of list)
    while current:
        # Print value with arrow if there's a next node
        print(current['val'], end=" -> " if current['next'] else "\n")
        
        # Move to next node by following the pointer
        # current['next'] gives us the next node dictionary
        # We assign it to current to "move forward"
        current = current['next']  # Follow the pointer to next node

print_linked_list(head=node1)  # Output: 1 -> 2 -> 3

# Alternative: Using helper function (cleaner!)
node1 = create_node(1)
node2 = create_node(2)
node3 = create_node(3)
node1['next'] = node2
node2['next'] = node3
print_linked_list(node1)  # Output: 1 -> 2 -> 3
````

````python
# Linked List Operations (using dictionaries)

# Helper function to create a node
def create_node(val, next_node=None):
    """Create a node dictionary"""
    return {'val': val, 'next': next_node}

# 1. Traversal - O(n) - Visit every node from head to end
def traverse(head):
    """
    Visit every node in the linked list.
    
    Time: O(n) - must visit each node once
    Space: O(1) - only uses one variable (current)
    
    How pointers work here:
    - current starts at head (first node)
    - current['next'] gives us the next node dictionary
    - We follow the pointer chain until we hit None
    """
    current = head  # Start at the head (first node)
    
    # Keep going until we reach None (end of list)
    while current:
        print(current['val'])  # Access value using dictionary key
        # Move to next node by following the pointer
        # This is the KEY: current['next'] is a pointer, we follow it!
        current = current['next']  # Follow pointer to next node

# 2. Search - O(n) - Look for a value in the list
def search(head, target):
    """
    Search for a value in linked list.
    
    Time: O(n) worst case - might need to check all nodes
    Space: O(1) - only uses one variable
    
    Why O(n)? Unlike arrays, we can't jump to middle. Must check one by one!
    """
    current = head  # Start at first node
    
    while current:
        if current['val'] == target:  # Found it! Access value using dictionary key
            return True
        
        # Not found yet - follow pointer to next node
        current = current['next']  # Move forward using pointer
    
    return False  # Reached end, not found

# 3. Insert at beginning - O(1) - Add new node as the head
def insert_at_head(head, val):
    """
    Insert new node at the beginning of list.
    
    Time: O(1) - constant time! Just create node and update pointer
    Space: O(1) - only creates one new node
    
    Visual:
    Before: head → [1|→] → [2|→] → None
    After:  head → [0|→] → [1|→] → [2|→] → None
    
    Steps:
    1. Create new node with value
    2. Point new node to old head (new_node['next'] = head)
    3. Return new node as new head
    """
    new_node = create_node(val)  # Create new node dictionary
    
    # Point new node's pointer to old head
    # This connects: new_node → old_head
    new_node['next'] = head
    
    # Return new node as the new head
    # Now new_node is the first node in the list!
    return new_node

# 4. Insert after a node - O(1) - Add new node after given node
def insert_after(node, val):
    """
    Insert new node after a specific node.
    
    Time: O(1) - constant time! Just update pointers
    Space: O(1) - only creates one new node
    
    Visual:
    Before: node → [A|→] → [B|→] → None
    After:  node → [A|→] → [new|→] → [B|→] → None
    
    Steps:
    1. Create new node
    2. Save what node currently points to (node['next'])
    3. Point new node to that saved node
    4. Point original node to new node
    """
    new_node = create_node(val)  # Create new node dictionary
    
    # Save what the original node points to
    # We need this because we're inserting BETWEEN node and node['next']
    new_node['next'] = node['next']  # New node points to what old node pointed to
    
    # Now point original node to new node
    # This creates: node → new_node → (what node['next'] used to point to)
    node['next'] = new_node

# 5. Delete node (given node to delete) - O(1) if we have the node
def delete_node(node):
    """
    Delete a node by copying next node's value and skipping it.
    
    Time: O(1) - constant time! Just copy value and update pointer
    Space: O(1) - no extra space needed
    
    Visual:
    Before: ... → [A|→] → [B|→] → [C|→] → ...
    After:  ... → [B|→] → [C|→] → ...
    
    Trick: Instead of deleting current node (hard!), we:
    1. Copy next node's value to current node
    2. Skip next node by pointing current['next'] to next['next']
    
    Note: This doesn't work if node is the last node (no next node to copy from)
    """
    if node['next']:  # Make sure there's a next node
        # Copy value from next node to current node
        node['val'] = node['next']['val']
        
        # Skip next node by pointing current['next'] to next['next']
        # This "deletes" next node by removing the pointer to it
        node['next'] = node['next']['next']

# Test
# Create linked list: 1 -> 2 -> 3
head = create_node(1)
head['next'] = create_node(2)
head['next']['next'] = create_node(3)

print("Original:")
print_linked_list(head)

head = insert_at_head(head, 0)
print("After inserting 0 at head:")
print_linked_list(head)

insert_after(head['next'], 1.5)  # Insert after first node
print("After inserting 1.5 after first node:")
print_linked_list(head)
````

````python
# Common Linked List Patterns

# Pattern 1: Two Pointers (Fast & Slow)
# Find middle of linked list
def find_middle(head):
    """Find middle node using two pointers"""
    slow = fast = head
    while fast and fast['next']:  # Access 'next' using dictionary key
        slow = slow['next']
        fast = fast['next']['next']
    return slow

# Pattern 2: Reverse Linked List
# Step-by-step visualization:

# Initial: 1 → 2 → 3 → 4 → None
#          ↑
#        head, current
#        prev = None

# Step 1: Save next, reverse link, move pointers
#  1 → 2 → 3 → 4 → None
#  ↑   ↑
# prev current
#  None ← 1  2 → 3 → 4 (link reversed!)

# Step 2: Continue...
#  None ← 1 ← 2  3 → 4
#         ↑   ↑
#        prev current

# Step 3: Continue...
#  None ← 1 ← 2 ← 3  4 → None
#              ↑   ↑
#            prev current

# Step 4: Final step
#  None ← 1 ← 2 ← 3 ← 4
#                  ↑   ↑
#                prev current (None)
# Return prev (new head = 4)

def reverse_linked_list(head):
    """Reverse a linked list"""
    prev = None
    current = head
    
    while current:
        next_node = current['next']  # Save next (before we lose it!) - use dictionary key
        current['next'] = prev        # Reverse the link
        prev = current                 # Move prev forward
        current = next_node            # Move current forward
    
    return prev  # New head

# Test
head = create_node(1)
head['next'] = create_node(2)
head['next']['next'] = create_node(3)
head['next']['next']['next'] = create_node(4)

print("Original:")
print_linked_list(head)

middle = find_middle(head)
print(f"Middle: {middle['val']}")  # Access value using dictionary key

reversed_head = reverse_linked_list(head)
print("Reversed:")
print_linked_list(reversed_head)
````

````python
# Pattern 3: Detect Cycle (Floyd's Cycle Detection)
# What is a cycle? When a node points back to a previous node, creating a loop.
# Example: 1 → 2 → 3 → 2 → 3 → 2... (3 points back to 2, creating infinite loop)
# Normal list: 1 → 2 → 3 → None (ends at None, no cycle)
# Why detect cycles? They cause infinite loops and crashes!

def has_cycle(head):
    """
    Use two pointers: slow moves 1 step, fast moves 2 steps
    If there's a cycle, they will meet (like runners on a circular track)
    """
    if not head or not head['next']:  # Access 'next' using dictionary key
        return False
    
    slow = head
    fast = head['next']
    
    while fast and fast['next']:
        if slow == fast:  # Pointers meet - cycle detected!
            return True
        slow = slow['next']
        fast = fast['next']['next']
    
    return False

# Test with cycle
node1 = create_node(1)
node2 = create_node(2)
node3 = create_node(3)
node1['next'] = node2
node2['next'] = node3
node3['next'] = node2  # Cycle! node3 points back to node2

print(f"Has cycle: {has_cycle(node1)}")

# Test without cycle
node1 = create_node(1)
node2 = create_node(2)
node3 = create_node(3)
node1['next'] = node2
node2['next'] = node3
# node3['next'] is None - no cycle

print(f"Has cycle: {has_cycle(node1)}")
````

```python
def remove_nth_from_end(head, n):
    """Remove nth node from end using dictionaries"""
    dummy = create_node(0)
    dummy['next'] = head
    
    first = second = dummy
    
    # Move first pointer n+1 steps ahead
    for _ in range(n + 1):
        first = first['next']
    
    # Move both until first reaches end
    while first:
        first = first['next']
        second = second['next']
    
    # Remove node
    second['next'] = second['next']['next']
    
    return dummy['next']
```

```python
def merge_two_lists(list1, list2):

    current = dummy

        if list1['val'] <= list2['val']:

            list1 = list1['next']

            current['next'] = list2
            list2 = list2['next']
        current = current['next']
    
    # Attach remaining
    current['next'] = list1 if list1 else list2
    
    return dummy['next']
```
