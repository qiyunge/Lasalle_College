## Part 1: Arrays - Deep Dive

### Array Operations & Patterns

Arrays are the most fundamental data structure. Let's master them!

````python
# Basic Array Operations
arr = [1, 2, 3, 4, 5]


# Access by index (O(1))
print(f"First element: {arr[0]}")
print(f"Last element: {arr[-1]}")
print(f"Middle element: {arr[len(arr)//2]}")

# Slicing (O(k) where k is slice size)
print(f"First 3: {arr[:3]}")
print(f"Last 2: {arr[-2:]}")
print(f"Middle: {arr[1:4]}")
print(f"Reversed: {arr[::-1]}")

# Common operations
arr.append(6)        # O(1) amortized
arr.insert(0, 0)    # O(n) - shifts elements
arr.pop()            # O(1) - removes last
arr.pop(0)           # O(n) - shifts elements
print(f"After operations: {arr}")
````

````python
# Traversal Patterns
arr = [10, 20, 30, 40, 50]

# Forward traversal
print("Forward:")
for i in range(len(arr)):
    print(f"Index {i}: {arr[i]}")

# Backward traversal
print("\nBackward:")
for i in range(len(arr) - 1, -1, -1):
    print(f"Index {i}: {arr[i]}")

"""ArithmeticError
"""

# Two pointers (from both ends)
print("\nTwo pointers from ends:")
left, right = 0, len(arr) - 1
while left < right:
    print(f"Left: {arr[left]}, Right: {arr[right]}")
    left += 1
    right -= 1
````

````python
# Finding elements
arr = [3, 7, 1, 9, 2, 5]

# Linear search - O(n)
def linear_search(arr, target):
    for i, num in enumerate(arr):
        if num == target:
            return i
    return -1

print(f"Index of 9: {linear_search(arr, 9)}")
print(f"Index of 10: {linear_search(arr, 10)}")

# Finding max/min - O(n)
print(f"Max: {max(arr)}, Min: {min(arr)}")
print(f"Max index: {arr.index(max(arr))}")
````

## Part 2: Strings

### String Manipulation

Strings are essentially arrays of characters. Many array techniques apply!

````python
# String Basics
s = "Hello World"

# Access characters (like arrays)
print(f"First char: {s[0]}")
print(f"Last char: {s[-1]}")
print(f"Substring: {s[0:5]}")

# String methods
print(f"Length: {len(s)}")
print(f"Uppercase: {s.upper()}")
print(f"Lowercase: {s.lower()}")
print(f"Split: {s.split()}")
print(f"Replace: {s.replace('World', 'Python')}")
print(f"Count 'l': {s.count('l')}")
print(f"Find 'World': {s.find('World')}")
````

````python
# String traversal
s = "python"

# Character by character
for char in s:
    print(char, end=" ")
print()

# With index
for i, char in enumerate(s):
    print(f"Index {i}: {char}")

# Reverse string
reversed_s = s[::-1]
print(f"Reversed: {reversed_s}")
````

````python
# Common string patterns

# Check if palindrome
def is_palindrome(s):
    s = s.lower().replace(" ", "")  # Normalize
    return s == s[::-1]

print(f"'racecar' is palindrome: {is_palindrome('racecar')}")
print(f"'hello' is palindrome: {is_palindrome('hello')}")

# Count character frequency
def count_chars(s):
    freq = {}
    for char in s:
        freq[char] = freq.get(char, 0) + 1
    return freq

print(f"Frequency: {count_chars('hello')}")
````

## Part 3: Hash Maps (Dictionaries)

### Understanding Hash Maps

**Hash Map** (Dictionary in Python) = Fast key-value storage with O(1) average operations!

**Think of it like:** A phone book - you know the name (key), instantly find the number (value)!

**How it works (Step-by-Step):**
```
Step 1: Store "Alice" → 95
  "Alice" → Hash Function → Index 2 → Store 95 at index 2

Step 2: Store "Bob" → 87
  "Bob" → Hash Function → Index 0 → Store 87 at index 0

Step 3: Lookup "Alice"
  "Alice" → Hash Function → Index 2 → Return 95 (O(1)!)
```

**Visual Example:**
```
Hash Map: {"Alice": 95, "Bob": 87, "Charlie": 92}
         ↓
    [Index]  [Key]      [Value]
      0      "Bob"   →   87
      1      None    →   None
      2      "Alice" →   95
      3      "Charlie" → 92
```

**Why O(1)?**
- Hash function instantly calculates index from key
- Direct array access = constant time
- No need to search through all keys!

**Key Concept:** Hash function converts key → array index for instant access!

**What is a hash function?** It's like a magic box that takes a name and gives you a box number!

**Simple Example:**
- You want to store "Alice" → 95
- Hash function: "Alice" → Box #2
- You put 95 in Box #2
- Later, when you ask "Where is Alice's data?"
- Hash function: "Alice" → Box #2 (same box!)
- You instantly go to Box #2 and get 95

**Why is this fast?** Instead of checking every box (Box 0, Box 1, Box 2...), the hash function tells you exactly which box to look in! Python does this automatically for you!

````python
# Hash Map Basics
# Create a hash map (dictionary) to store student grades
# Format: {key: value} where key is student name, value is grade
student_grades = {"Alice": 95, "Bob": 87, "Charlie": 92}

# All these operations are O(1) average time - super fast!

# 1. Lookup: Get value by key
# How it works: Hash function converts "Alice" → index → get value instantly
print(f"Alice's grade: {student_grades['Alice']}")  # O(1) - instant lookup!

# 2. Insert: Add new key-value pair
# How it works: Hash function converts "Diana" → index → store value at that index
student_grades["Diana"] = 88  # O(1) - instant insert!

# 3. Update: Change value for existing key
# How it works: Same as insert - hash function finds index, updates value there
student_grades["Bob"] = 90    # O(1) - instant update!

# 4. Delete: Remove key-value pair
# How it works: Hash function finds index, marks it as empty
del student_grades["Charlie"]  # O(1) - instant delete!

print(f"Updated: {student_grades}")

# 5. Check existence: Is key in dictionary?
# How it works: Hash function checks if key exists at calculated index
print(f"'Alice' in dict: {'Alice' in student_grades}")  # O(1) - instant check!

# 6. Get with default: Get value or return default if key doesn't exist
# Why useful? Prevents KeyError if key doesn't exist
# .get(key, default) returns default value if key not found
print(f"Get with default: {student_grades.get('Eve', 'Not found')}")  # O(1)
````

### The Two Sum Problem Pattern

**Problem:** Find two numbers that add up to target.

**The Idea:** Instead of checking all pairs (O(n²)), store what we've seen and check if complement exists!

**Step-by-Step Walkthrough:**
```
Array: [2, 7, 11, 15], target = 9
seen = {}  (stores: value → index)

i=0, num=2:
  complement = 9 - 2 = 7
  Is 7 in seen? No → Add 2 to seen: {2: 0}

i=1, num=7:
  complement = 9 - 7 = 2
  Is 2 in seen? Yes! → Found pair: [seen[2], 1] = [0, 1] ✓
  
Answer: indices [0, 1] → values [2, 7] = 9
```

**Why Hash Map?**
- Without hash map: Check every pair → O(n²) time
- With hash map: One pass, O(1) lookup → O(n) time!

**Key Insight:** For each number, we need its complement (target - num). Store numbers as we go!

**What is "complement"?** 
- The complement is "the other number we need to complete the pair"
- If target = 9 and we see num = 2, then complement = 9 - 2 = 7
- We're looking for 7 because 2 + 7 = 9 ✓
- If we've already seen 7 in our hash map, we found our pair!
- Think of it like: "I have 2, I need 7 to make 9. Have I seen 7 before?"

````python
# Common hash map patterns

# Pattern 1: Frequency counting
# Count how many times each value appears
def count_frequency(arr):
    """
    Count frequency of each number in array.
    
    Example: [1, 2, 2, 3, 3, 3] → {1: 1, 2: 2, 3: 3}
    
    Why use .get()? 
    - If key doesn't exist, .get(key, 0) returns 0 (default)
    - Then we add 1 to get frequency = 1
    - If key exists, .get(key, 0) returns current count
    - Then we add 1 to increment frequency
    """
    freq = {}  # Hash map: number → count
    
    for num in arr:
        # Get current count (0 if first time seeing this number)
        # Then add 1 to increment the count
        freq[num] = freq.get(num, 0) + 1
    
    return freq

arr = [1, 2, 2, 3, 3, 3, 4]
print(f"Frequency: {count_frequency(arr)}")
# Output: {1: 1, 2: 2, 3: 3, 4: 1}
# Meaning: 1 appears 1 time, 2 appears 2 times, 3 appears 3 times, 4 appears 1 time

# Pattern 2: Index mapping
# Map each value to its index (or last index if duplicates)
def create_index_map(arr):
    """
    Create mapping from value to its index.
    
    Example: ['a', 'b', 'c'] → {'a': 0, 'b': 1, 'c': 2}
    
    Why useful? O(1) lookup to find index of a value!
    Without hash map: O(n) linear search to find index
    """
    index_map = {}  # Hash map: value → index
    
    # enumerate gives us (index, value) pairs
    for i, val in enumerate(arr):
        # Store: value → its index
        # If value appears multiple times, this stores the LAST index
        index_map[val] = i
    
    return index_map

print(f"Index map: {create_index_map(['a', 'b', 'c'])}")
# Output: {'a': 0, 'b': 1, 'c': 2}
# Now we can do: index_map['b'] → instantly get 1 (O(1) instead of O(n)!)
````

````python
# The Two Sum Problem Pattern
# Find two numbers that add up to target

def two_sum(arr, target):
    """
    Find two numbers that add up to target.
    
    Time: O(n) - single pass through array
    Space: O(n) - hash map stores up to n elements
    
    Without hash map: O(n²) - would need nested loops to check all pairs
    
    Strategy:
    1. For each number, calculate what we need (complement = target - num)
    2. Check if we've seen the complement before (O(1) lookup!)
    3. If yes, we found our pair!
    4. If no, store current number for future lookups
    
    Example: arr = [2, 7, 11, 15], target = 9
    - i=0, num=2: complement=7, not in seen → store {2: 0}
    - i=1, num=7: complement=2, IS in seen! → return [0, 1] ✓
    """
    seen = {}  # Hash map: value → index (stores numbers we've seen so far)
    
    # Go through array once
    for i, num in enumerate(arr):
        # Calculate what number we need to pair with current number
        complement = target - num
        
        # Check if we've seen the complement before (O(1) lookup!)
        if complement in seen:
            # Found it! Return indices of complement and current number
            return [seen[complement], i]
        
        # Haven't seen complement yet - store current number for future
        # Store: number → its index (so we can return indices later)
        seen[num] = i
    
    # No pair found
    return []

# Test
arr = [2, 7, 11, 15]
target = 9
result = two_sum(arr, target)
print(f"Two sum indices: {result}")
print(f"Values: {arr[result[0]]}, {arr[result[1]]}")
````

## Part 4: Sets

### Set Operations

**Sets** are like hash maps but only store keys (no values). Perfect for:
- Checking membership (O(1))
- Removing duplicates
- Set operations (union, intersection, difference)

````python
# Set Basics
numbers = {1, 2, 3, 4, 5}

# O(1) membership check
print(f"3 in set: {3 in numbers}")
print(f"10 in set: {10 in numbers}")

# Add/remove
numbers.add(6)      # O(1)
numbers.remove(1)   # O(1)
print(f"After operations: {numbers}")
````

````python
# Set Operations
set_a = {1, 2, 3, 4, 5}
set_b = {4, 5, 6, 7, 8}

print(f"Set A: {set_a}")
print(f"Set B: {set_b}")
print(f"Union (A | B): {set_a | set_b}")
print(f"Intersection (A & B): {set_a & set_b}")
print(f"Difference (A - B): {set_a - set_b}")
print(f"Symmetric Difference (A ^ B): {set_a ^ set_b}")  # (A-B) & (B-A) = {1,2,3} + {6, 7, 8} = {1,2,3,6,7,8}
````

````python
# Common set patterns

# Remove duplicates from list
arr = [1, 2, 2, 3, 3, 3, 4]
unique = list(set(arr))
print(f"Unique: {unique}")

# Check for duplicates
def has_duplicates(arr):
    return len(arr) != len(set(arr))

print(f"Has duplicates: {has_duplicates([1, 2, 3])}")
print(f"Has duplicates: {has_duplicates([1, 2, 2, 3])}")
````

## Part 5: When to Use What?

| Data Structure | Use When | Time Complexity |
|----------------|----------|----------------|
| **Array** | Need ordered data, index access | Access: O(1), Search: O(n) |
| **Hash Map** | Need key-value pairs, fast lookup | All ops: O(1) avg |
| **Set** | Need uniqueness, membership checks | All ops: O(1) avg |

**Decision Tree:**
- Need fast lookup by key? → Hash Map
- Need to check if element exists? → Set
- Need ordered/indexed data? → Array
- Need to count frequencies? → Hash Map

```python
def group_anagrams(strs):
    groups = {}
    for s in strs:
        key = ''.join(sorted(s))
        if key not in groups:
            groups[key] = []
        groups[key].append(s)
    return list(groups.values())
```

### Solution 4: Longest Substring (Sliding Window)
```python
def length_of_longest_substring(s):
    char_set = set()
    left = 0
    max_len = 0
    
    for right in range(len(s)):
        while s[right] in char_set:
            char_set.remove(s[left])
            left += 1
        char_set.add(s[right])
        max_len = max(max_len, right - left + 1)
    
    return max_len
```
