```python
import pandas as pd
import matplotlib.pyplot as plt

DATA_URL = "./data/Descriptive-statistics-HomePrice.xlsx"
SHEET_NAMES = "Data"

re_pd = pd.read_excel(DATA_URL, sheet_name=SHEET_NAMES)
# re_pd.head()

bins = 10

price_bins = pd.cut(re_pd["Price"], bins)
price_bins.head()
# counts = price_bins.value_counts().sort_index()
# ax = counts.plot.bar(figsize=(6, 4))
# plt.tight_layout()


re_pd["Price"].plot.hist(bins=bins, alpha=0.5, figsize=(6, 4))
plt.tight_layout()
plt.show()


```

```python
#q_4
# plt.scatter(re_pd["Price"], re_pd["Area (ft.)"])

re_pd.plot.scatter(x="Price", y="Area (ft.)", figsize=(6, 4))
corr_p_a =re_pd[["Price","Area (ft.)"]].corr()
print(corr_p_a)
plt.tight_layout()
plt.show()
```

```python

x = re_pd.loc[re_pd["Mortgage"] == "No"]["Price"]
y = re_pd.loc[re_pd["Mortgage"] == "Yes"]["Price"]

x.plot.hist(bins=20, alpha=0.5, label='No Mortgage', figsize=(6,4))
y.plot.hist(bins=20, alpha=0.5, label='Yes Mortgage', figsize=(6,4))
plt.legend()
plt.tight_layout()
plt.show()

# bins = 20
# plt.hist(x,bins =bins, alpha=0.5, label='No Mortgage')
# plt.hist(y,bins =bins, alpha=0.5, label='Yes Mortgage')
# plt.legend()
# plt.tight_layout()
# plt.show()
```

```python
female = re_pd[re_pd["Gender"]=="F"]["Price"]
male = re_pd[re_pd["Gender"]=="M"]["Price"] 
bins = 20
plt.hist(male,bins =bins, alpha=0.5, label='Male')  
plt.hist(female,bins =bins, alpha=0.5, label='Female')
plt.legend()
plt.tight_layout()
plt.show()
```
