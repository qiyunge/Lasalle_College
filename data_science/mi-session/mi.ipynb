```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = pd.read_excel('G5_midterm_FR.xlsx', sheet_name='HR data' ,header=3)
data.dropna(inplace=True,axis =1, how='all')
```

```python
data.head()
data.info()
```

```python
data['Salary'].plot.hist(bins=20)
plt.xlabel('Salary')
plt.title('Distribution of Salary')
plt.show()  
```

```python
data.plot.scatter(x='Age', y='Salary')
plt.title('Age vs Salary')

```

```python
data["Salary"].describe()
data["Salary"].mode()
```

```python
fig,axes = plt.subplots(1, 2, figsize=(12, 5))

data.pivot_table(index='Department', values='Salary', aggfunc='mean').plot.bar(ax=axes[0])
axes[0].set_title('Average Salary by Department')
axes[0].set_ylabel('Average Salary')
axes[0].set_xlabel('Department')
axes[0].set_xticklabels(data["Department"].unique(), rotation=45)       

data.pivot_table(index='Ethnicity', values='Salary', aggfunc='mean').plot.bar(ax=axes[1])
axes[1].set_title('Average Salary by Ethnicity')
axes[1].set_ylabel('Average Salary')
axes[1].set_xlabel(' Ethnicity')
axes[1].set_xticklabels(data["Ethnicity"].unique(), rotation=45)
plt.show()
data["Ethnicity"].unique()
```
