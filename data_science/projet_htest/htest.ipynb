```python
import pandas as pd
from scipy.stats import chi2_contingency, f_oneway
```

```python
census = pd.read_excel("../data/Assignment-census_income.xlsx", sheet_name="census_income")
census.head()
```

```python
census.info()
```

```python
census.describe()
```

```python
# ho: There is race gap based on education 
# h1: There is a race gap based on education

race_education = pd.crosstab(census["education"], census["race"])
race_education
```

```python
res = chi2_contingency(race_education)
chi2, p, dof, expected = res
print("Chi2 Statistic:", chi2)
print("P-Value:", p)
print("Degrees of Freedom:", dof)
print("Expected Frequencies:", expected)
print("Chi2:", chi2)
```

```python

census = census.applymap(lambda x: x.strip() if x.dtype == "str" else x)
census_Federal_gov = census[census["workclass"] == " Federal-gov"]["education.num"]
# print(census_Federal_gov)
census_Local_gov = census[census["workclass"] == " Local-gov"]["education.num"]
# print(census_Federal_gov)

census_never_worked = census[census["workclass"] == " Never-worked"]["education.num"]
census_private = census[census["workclass"] == " Private"]["education.num"]
census_self_emp_inc = census[census["workclass"] == " Self-emp-inc"]["education.num"]
census_self_emp_not_inc = census[census["workclass"] == " Self-emp-not-inc"]["education.num"]
census_state_gov = census[census["workclass"] == " State-gov"]["education.num"]
census_without_pay = census[census["workclass"] == " Without-pay"]["education.num"]
census_question_mark = census[census["workclass"] == " ?"]["education.num"]  

f_statistic, p_value = f_oneway(census_Federal_gov, census_Local_gov, census_never_worked, census_private, census_self_emp_inc, census_self_emp_not_inc, census_state_gov, census_without_pay, census_question_mark)
p_value
```
